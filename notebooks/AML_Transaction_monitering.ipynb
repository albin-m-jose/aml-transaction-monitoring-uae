```python
# =============================================================================
# UAE AML SYNTHETIC DATA GENERATOR
# For Transaction Monitoring & AML Detection Projects
# =============================================================================

import os
import warnings
from datetime import datetime, timedelta

import numpy as np
import pandas as pd
from faker import Faker

warnings.filterwarnings("ignore")

# --- OPTIONAL PROGRESS BAR ---------------------------------------------------
try:
    from tqdm.notebook import tqdm
except ImportError:

    def tqdm(iterable, **kwargs):
        return iterable


# =============================================================================
# CONFIGURATION
# =============================================================================

SEED = 42
np.random.seed(SEED)
random_state = np.random.RandomState(SEED)
fake = Faker()
fake.seed_instance(SEED)

NUM_CUSTOMERS = 10000  # Scale up as needed
TRANS_MONTHS = 12
START_DATE = datetime(2023, 1, 1)
END_DATE = START_DATE + timedelta(days=30 * TRANS_MONTHS)

OUTPUT_DIR = "./aml_synthetic_uae"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# =============================================================================
# DISTRIBUTIONS & MAPPINGS
# =============================================================================

# --- Customer Segment Mix ----------------------------------------------------
SEGMENTS = {
    "retail": 0.78,
    "sme": 0.12,
    "corporate": 0.06,
    "dnfbp": 0.03,
    "vasp": 0.01,
}

# --- Retail Nationalities (expat-heavy) --------------------------------------
RETAIL_NATS = [
    "India",
    "Pakistan",
    "Philippines",
    "Bangladesh",
    "Egypt",
    "Nepal",
    "Sri Lanka",
    "United Kingdom",
    "Jordan",
    "Sudan",
    "United Arab Emirates",
    "Lebanon",
    "Russia",
    "Nigeria",
    "Iran",
    "China",
    "Other",
]
RETAIL_NAT_WEIGHTS = [
    28,
    13,
    7,
    8,
    6,
    4,
    4,
    4,
    3,
    3,
    16,
    3,
    2,
    1.5,
    1,
    1.5,
    2,
]

# --- Age & Gender ------------------------------------------------------------
AGE_BINS = [18, 25, 35, 45, 55, 65, 80]
AGE_WEIGHTS = [0.12, 0.35, 0.28, 0.18, 0.05, 0.02]
GENDER_WEIGHTS = [0.68, 0.32]  # Male, Female

# --- Emirates ----------------------------------------------------------------
EMIRATES = [
    "Dubai",
    "Abu Dhabi",
    "Sharjah",
    "Ajman",
    "Ras Al Khaimah",
    "Fujairah",
    "Umm Al Quwain",
]
EMIRATE_WEIGHTS = [0.55, 0.25, 0.08, 0.04, 0.03, 0.03, 0.02]

# --- Occupations by Nationality Cluster --------------------------------------
OCCUPATIONS_LOW = [
    "Construction Laborer",
    "Domestic Worker",
    "Delivery Driver",
    "Taxi Driver",
    "Retail Shop Worker",
    "Security Guard",
    "Waiter",
    "Salon Worker",
]
OCCUPATIONS_MID = [
    "Nurse",
    "Teacher",
    "Accountant",
    "Sales Executive",
    "IT Support Technician",
    "Shop Owner",
    "Admin Assistant",
]
OCCUPATIONS_HIGH = [
    "Engineer",
    "Doctor",
    "Business Owner",
    "Finance Manager",
    "IT Consultant",
]

OCCUPATION_INCOME = {
    "Construction Laborer": (1500, 4000),
    "Domestic Worker": (1200, 3500),
    "Delivery Driver": (2000, 5000),
    "Taxi Driver": (2500, 6000),
    "Security Guard": (1800, 4000),
    "Retail Shop Worker": (2000, 5000),
    "Waiter": (2000, 4500),
    "Salon Worker": (2500, 5500),
    "Nurse": (6000, 15000),
    "Teacher": (7000, 18000),
    "Accountant": (8000, 20000),
    "Sales Executive": (7000, 25000),
    "IT Support Technician": (8000, 18000),
    "Shop Owner": (8000, 40000),
    "Engineer": (15000, 45000),
    "Doctor": (25000, 80000),
    "Business Owner": (15000, 100000),
    "Finance Manager": (20000, 60000),
    "IT Consultant": (18000, 50000),
    "Government Official": (15000, 50000),
    "Admin Assistant": (5000, 12000),
}

# --- SME / Corporate / DNFBP / VASP ------------------------------------------
SME_TYPES = [
    "General Trading",
    "Restaurant / F&B",
    "Construction / Contracting",
    "Consultancy",
    "E-commerce",
    "Real Estate Brokerage",
    "Gold / Jewelry Trading",
    "Logistics / Shipping",
    "Freelance",
]
SME_TYPE_WEIGHTS = [0.22, 0.15, 0.12, 0.14, 0.10, 0.08, 0.05, 0.07, 0.07]

CORP_INDUSTRY = [
    "Real Estate Developer",
    "Government-Linked",
    "MNC Regional HQ",
    "Bank / FI",
    "Airline / Logistics",
    "Oil / Energy",
    "Hospitality",
    "Healthcare",
    "Education",
]
CORP_INDUSTRY_WEIGHTS = [
    0.20,
    0.18,
    0.15,
    0.10,
    0.08,
    0.07,
    0.12,
    0.05,
    0.05,
]

DNFBP_TYPES = [
    "Real Estate Broker",
    "Gold Dealer",
    "Luxury Car Dealer",
    "Corporate Service Provider",
    "Lawyer",
    "Accountant",
]
DNFBP_WEIGHTS = [0.35, 0.25, 0.15, 0.15, 0.05, 0.05]

VASP_TYPES = ["Crypto Exchange", "OTC Desk", "Custody Provider", "Broker / ATM"]
VASP_WEIGHTS = [0.40, 0.30, 0.20, 0.10]

# --- Account Types -----------------------------------------------------------
RETAIL_ACC_TYPES = {
    "low": {"Basic Savings": 0.6, "Salary Account": 0.4},
    "mid": {
        "Current Account": 0.5,
        "Basic Savings": 0.3,
        "Fixed Deposit": 0.2,
    },
    "high": {
        "Current Account": 0.4,
        "Wealth / Priority": 0.35,
        "Multi-currency": 0.25,
    },
}

# --- Suspicious Pattern Config -----------------------------------------------
SUSPICIOUS_RATE = 0.035  # 3.5% of customers are "bad actors"
SUSPICION_TYPES = [
    "structuring",
    "rapid_movement",
    "layering",
    "pep_unusual",
    "trade_misinvoicing",
    "hawala",
]

# --- Cross-border corridors ----------------------------------------------------
CORRIDORS = {
    "India": 0.22,
    "Pakistan": 0.12,
    "Philippines": 0.08,
    "UK": 0.07,
    "USA": 0.06,
    "China": 0.05,
    "Saudi Arabia": 0.08,
    "Other GCC": 0.10,
    "Offshore": 0.08,
    "High-Risk": 0.05,
    "Africa": 0.05,
    "Other": 0.04,
}

# =============================================================================
# HELPER FUNCTIONS
# =============================================================================


def random_date(start, end):
    """Return random datetime between start and end."""
    delta = end - start
    return start + timedelta(
        seconds=random_state.randint(0, max(1, int(delta.total_seconds())))
    )


def assign_emirate(nationality, income):
    """Emirate selection based on demographics."""
    if nationality in ["Bangladesh", "Nepal", "Pakistan"] and income < 5000:
        return random_state.choice(
            ["Dubai", "Sharjah", "Ajman"], p=[0.5, 0.3, 0.2]
        )
    elif income > 50000:
        return random_state.choice(["Dubai", "Abu Dhabi"], p=[0.7, 0.3])
    return random_state.choice(EMIRATES, p=EMIRATE_WEIGHTS)


def assign_occupation(nationality, age):
    """Occupation correlated with nationality (Dubai labor market reality)."""
    if nationality in ["Bangladesh", "Nepal"]:
        occs = OCCUPATIONS_LOW
        weights = [0.50, 0.10, 0.15, 0.10, 0.05, 0.05, 0.03, 0.02]
    elif nationality == "Philippines":
        occs = [
            "Domestic Worker",
            "Nurse",
            "Salon Worker",
            "Retail Shop Worker",
            "Teacher",
        ]
        weights = [0.30, 0.25, 0.15, 0.15, 0.15]
    elif nationality in ["India", "Pakistan"]:
        occs = OCCUPATIONS_LOW + OCCUPATIONS_MID + OCCUPATIONS_HIGH
        # 8 low + 7 mid + 5 high = 20 occupations
        weights = [
            0.14,
            0.05,
            0.08,
            0.08,
            0.05,
            0.07,
            0.05,
            0.04,  # LOW (8)
            0.07,
            0.06,
            0.06,
            0.05,
            0.05,
            0.04,
            0.04,  # MID (7)
            0.02,
            0.02,
            0.03,
            0.02,
            0.02,  # HIGH (5)
        ]
    elif nationality == "United Arab Emirates":
        occs = [
            "Government Official",
            "Business Owner",
            "Finance Manager",
            "Engineer",
            "Doctor",
            "Teacher",
            "Admin Assistant",
        ]
        weights = [0.30, 0.15, 0.15, 0.15, 0.10, 0.10, 0.05]
    elif nationality in ["United Kingdom", "Russia"]:
        occs = [
            "Finance Manager",
            "IT Consultant",
            "Business Owner",
            "Engineer",
            "Doctor",
        ]
        weights = [0.25, 0.20, 0.20, 0.20, 0.15]
    elif nationality == "Egypt":
        occs = [
            "Accountant",
            "Teacher",
            "Sales Executive",
            "Engineer",
            "Nurse",
            "Admin Assistant",
        ]
        weights = [0.20, 0.20, 0.20, 0.20, 0.10, 0.10]
    else:
        occs = OCCUPATIONS_MID + OCCUPATIONS_HIGH
        # 7 mid + 5 high = 12 occupations
        weights = [
            0.10,
            0.10,
            0.10,
            0.10,
            0.10,
            0.10,
            0.08,  # MID (7)
            0.07,
            0.07,
            0.06,
            0.06,
            0.06,  # HIGH (5)
        ]

    return random_state.choice(occs, p=np.array(weights) / sum(weights))


def income_from_occupation(occ):
    lo, hi = OCCUPATION_INCOME.get(occ, (5000, 15000))
    return int(random_state.lognormal(np.log((lo + hi) / 2), 0.3))


def kyc_risk_rating(nationality, occupation, income, pep_flag, is_sme=False):
    score = 0
    if pep_flag:
        score += 40
    if nationality in ["Iran", "Syria", "North Korea", "Afghanistan"]:
        score += 35
    if occupation in ["Business Owner", "Shop Owner", "Gold Dealer"]:
        score += 15
    if income > 80000:
        score += 10
    if is_sme:
        score += 10
    if nationality in ["Nigeria", "Russia"]:
        score += 10

    if score >= 40:
        return "High"
    if score >= 20:
        return "Medium"
    return "Low"


# =============================================================================
# 1. GENERATE CUSTOMERS
# =============================================================================

print("Generating customers...")
customers = []

seg_counts = {k: int(NUM_CUSTOMERS * v) for k, v in SEGMENTS.items()}
seg_counts["retail"] += NUM_CUSTOMERS - sum(seg_counts.values())

# --- RETAIL ---
for i in tqdm(range(seg_counts["retail"]), desc="Retail"):
    nat = random_state.choice(
        RETAIL_NATS, p=np.array(RETAIL_NAT_WEIGHTS) / sum(RETAIL_NAT_WEIGHTS)
    )
    age = int(random_state.choice([22, 30, 40, 50, 60, 70], p=AGE_WEIGHTS))
    gender = random_state.choice(["M", "F"], p=GENDER_WEIGHTS)

    if nat == "United Arab Emirates" and random_state.random() < 0.4:
        gender = "F"
    if nat == "Philippines" and random_state.random() < 0.6:
        gender = "F"

    occ = assign_occupation(nat, age)
    income = income_from_occupation(occ)

    pep_flag = random_state.random() < 0.018
    if pep_flag and nat != "United Arab Emirates":
        pep_flag = random_state.random() < 0.6

    cust = {
        "customer_id": f"CUST{100000 + i}",
        "segment": "retail",
        "customer_type": "individual",
        "name": fake.name_male() if gender == "M" else fake.name_female(),
        "nationality": nat,
        "gender": gender,
        "age": age,
        "occupation": occ,
        "monthly_income": income,
        "income_level": (
            "low"
            if income < 8000
            else ("mid" if income < 25000 else "high")
        ),
        "pep_flag": pep_flag,
        "pep_category": (
            random_state.choice(["domestic", "foreign", "international_org"])
            if pep_flag
            else None
        ),
        "sanctions_flag": random_state.random() < 0.0005,
        "adverse_media_flag": random_state.random() < 0.003,
        "high_risk_jurisdiction_flag": nat
        in ["Iran", "Syria", "Afghanistan", "North Korea"],
        "kyc_risk_rating": kyc_risk_rating(nat, occ, income, pep_flag),
        "account_open_date": random_date(
            START_DATE - timedelta(days=365 * 3),
            END_DATE - timedelta(days=180),
        ),
        "emirate": assign_emirate(nat, income),
        "visa_type": random_state.choice(
            ["Employment", "Investor", "Tourist", "Golden", "Student", "None"]
        ),
        "employment_type": random_state.choice(
            [
                "Salaried-Private",
                "Salaried-Govt",
                "Self-employed",
                "Business Owner",
                "Unemployed",
                "Student",
                "Retired",
            ],
            p=[0.55, 0.12, 0.15, 0.10, 0.03, 0.03, 0.02],
        ),
        "suspicion_label": "normal",
        "shell_company_risk_score": 0,
    }
    customers.append(cust)

# --- SME / FREELANCER ---
for i in tqdm(range(seg_counts["sme"]), desc="SME"):
    sme_type = random_state.choice(SME_TYPES, p=SME_TYPE_WEIGHTS)
    turnover = random_state.choice([
        random_state.lognormal(13, 1),
        random_state.lognormal(16, 0.5),
        random_state.lognormal(18, 0.8),
    ])
    turnover = max(50000, min(50000000, turnover))

    cust = {
        "customer_id": f"SME{200000 + i}",
        "segment": "sme",
        "customer_type": "entity",
        "name": fake.company(),
        "nationality": random_state.choice(
            ["India", "Pakistan", "UAE", "UK", "Lebanon", "China", "Russia", "Other"],
            p=[0.25, 0.10, 0.20, 0.10, 0.10, 0.10, 0.05, 0.10],
        ),
        "gender": None,
        "age": None,
        "occupation": sme_type,
        "monthly_income": None,
        "annual_turnover": int(turnover),
        "income_level": "business",
        "pep_flag": random_state.random() < 0.03,
        "pep_category": None,
        "sanctions_flag": random_state.random() < 0.001,
        "adverse_media_flag": random_state.random() < 0.005,
        "high_risk_jurisdiction_flag": False,
        "kyc_risk_rating": kyc_risk_rating(
            "Other", sme_type, turnover, False, is_sme=True
        ),
        "account_open_date": random_date(
            START_DATE - timedelta(days=365 * 2), END_DATE - timedelta(days=90)
        ),
        "emirate": random_state.choice(EMIRATES, p=EMIRATE_WEIGHTS),
        "visa_type": None,
        "employment_type": None,
        "suspicion_label": "normal",
        "shell_company_risk_score": (
            random_state.randint(0, 100)
            if random_state.random() < 0.03
            else random_state.randint(0, 30)
        ),
        "business_size_employees": random_state.choice(
            [1, 3, 10, 30, 100, 250], p=[0.20, 0.35, 0.30, 0.11, 0.03, 0.01]
        ),
        "legal_structure": random_state.choice(
            ["Sole Establishment", "LLC", "Free Zone", "Branch", "Civil"],
            p=[0.40, 0.25, 0.25, 0.07, 0.03],
        ),
    }
    customers.append(cust)

# --- CORPORATE ---
for i in tqdm(range(seg_counts["corporate"]), desc="Corporate"):
    industry = random_state.choice(CORP_INDUSTRY, p=CORP_INDUSTRY_WEIGHTS)
    turnover = random_state.lognormal(20, 1.5)
    turnover = max(1000000, min(500000000, turnover))

    cust = {
        "customer_id": f"CORP{300000 + i}",
        "segment": "corporate",
        "customer_type": "entity",
        "name": fake.company()
        + " "
        + random_state.choice(["Group", "LLC", "Holding", "Ltd", "FZE"]),
        "nationality": "United Arab Emirates",
        "gender": None,
        "age": None,
        "occupation": industry,
        "monthly_income": None,
        "annual_turnover": int(turnover),
        "income_level": "corporate",
        "pep_flag": random_state.random() < 0.08,
        "pep_category": (
            "domestic" if random_state.random() < 0.8 else "foreign"
        ),
        "sanctions_flag": False,
        "adverse_media_flag": random_state.random() < 0.01,
        "high_risk_jurisdiction_flag": False,
        "kyc_risk_rating": random_state.choice(
            ["Low", "Medium"], p=[0.7, 0.3]
        ),
        "account_open_date": random_date(
            START_DATE - timedelta(days=365 * 5),
            END_DATE - timedelta(days=365),
        ),
        "emirate": random_state.choice(["Dubai", "Abu Dhabi"], p=[0.6, 0.4]),
        "visa_type": None,
        "employment_type": None,
        "suspicion_label": "normal",
        "shell_company_risk_score": random_state.randint(0, 15),
    }
    customers.append(cust)

# --- DNFBP ---
for i in tqdm(range(seg_counts["dnfbp"]), desc="DNFBP"):
    dnfbp_type = random_state.choice(DNFBP_TYPES, p=DNFBP_WEIGHTS)
    cust = {
        "customer_id": f"DNFBP{400000 + i}",
        "segment": "dnfbp",
        "customer_type": "entity",
        "name": fake.company() + " " + dnfbp_type.split()[-1],
        "nationality": random_state.choice(
            ["UAE", "Lebanon", "India", "Iran", "UK", "Other"],
            p=[0.40, 0.20, 0.15, 0.10, 0.10, 0.05],
        ),
        "gender": None,
        "age": None,
        "occupation": dnfbp_type,
        "monthly_income": None,
        "annual_turnover": int(random_state.lognormal(16, 1.2)),
        "income_level": "business",
        "pep_flag": random_state.random() < 0.05,
        "pep_category": None,
        "sanctions_flag": random_state.random() < 0.002,
        "adverse_media_flag": random_state.random() < 0.02,
        "high_risk_jurisdiction_flag": False,
        "kyc_risk_rating": (
            "High"
            if dnfbp_type == "Gold Dealer"
            else random_state.choice(["Medium", "High"], p=[0.4, 0.6])
        ),
        "account_open_date": random_date(
            START_DATE - timedelta(days=365 * 3),
            END_DATE - timedelta(days=180),
        ),
        "emirate": "Dubai",
        "visa_type": None,
        "employment_type": None,
        "suspicion_label": "normal",
        "shell_company_risk_score": (
            random_state.randint(10, 60)
            if dnfbp_type == "Corporate Service Provider"
            else random_state.randint(0, 40)
        ),
    }
    customers.append(cust)

# --- VASP ---
for i in tqdm(range(seg_counts["vasp"]), desc="VASP"):
    vasp_type = random_state.choice(VASP_TYPES, p=VASP_WEIGHTS)
    cust = {
        "customer_id": f"VASP{500000 + i}",
        "segment": "vasp",
        "customer_type": "entity",
        "name": fake.company() + " Digital Assets",
        "nationality": random_state.choice(
            ["UAE", "USA", "UK", "Singapore", "Other"],
            p=[0.50, 0.15, 0.15, 0.10, 0.10],
        ),
        "gender": None,
        "age": None,
        "occupation": vasp_type,
        "monthly_income": None,
        "annual_turnover": int(random_state.lognormal(17, 1.5)),
        "income_level": "business",
        "pep_flag": False,
        "pep_category": None,
        "sanctions_flag": False,
        "adverse_media_flag": random_state.random() < 0.01,
        "high_risk_jurisdiction_flag": False,
        "kyc_risk_rating": "High",
        "account_open_date": random_date(
            START_DATE - timedelta(days=365 * 2), END_DATE - timedelta(days=90)
        ),
        "emirate": "Dubai",
        "visa_type": None,
        "employment_type": None,
        "suspicion_label": "normal",
        "shell_company_risk_score": random_state.randint(0, 30),
        "license_status": random_state.choice(
            ["Licensed", "Unlicensed"], p=[0.85, 0.15]
        ),
    }
    customers.append(cust)

df_customers = pd.DataFrame(customers)
print(f"Customers generated: {len(df_customers)}")

# =============================================================================
# 2. ASSIGN SUSPICIOUS LABELS TO SUBSET OF CUSTOMERS
# =============================================================================

print("Injecting suspicious actor labels...")
n_suspicious = int(len(df_customers) * SUSPICIOUS_RATE)
suspicious_idx = random_state.choice(
    df_customers.index, size=n_suspicious, replace=False
)

suspicion_draw = random_state.choice(
    SUSPICION_TYPES,
    size=n_suspicious,
    p=[0.25, 0.20, 0.20, 0.15, 0.12, 0.08],
)

for idx, stype in zip(suspicious_idx, suspicion_draw):
    seg = df_customers.loc[idx, "segment"]
    if seg == "retail" and stype in ["trade_misinvoicing"]:
        stype = random_state.choice(
            ["structuring", "rapid_movement", "layering", "hawala"]
        )
    if seg in ["corporate", "vasp"] and stype in ["structuring", "hawala"]:
        stype = random_state.choice(
            ["rapid_movement", "layering", "pep_unusual", "trade_misinvoicing"]
        )

    df_customers.at[idx, "suspicion_label"] = stype
    df_customers.at[idx, "kyc_risk_rating"] = "High"

print(
    f"Suspicious customers: {(df_customers['suspicion_label'] != 'normal').sum()}"
)

# =============================================================================
# 3. GENERATE ACCOUNTS
# =============================================================================

print("Generating accounts...")
accounts = []

for _, cust in tqdm(
    df_customers.iterrows(), total=len(df_customers), desc="Accounts"
):
    n_accounts = random_state.choice([1, 2, 3], p=[0.65, 0.25, 0.10])
    if cust["segment"] in ["sme", "corporate", "dnfbp"]:
        n_accounts = random_state.choice(
            [1, 2, 3, 4], p=[0.30, 0.30, 0.25, 0.15]
        )

    for acc_i in range(n_accounts):
        if cust["segment"] == "retail":
            inc_level = cust["income_level"]
            if inc_level == "low":
                acc_type = random_state.choice(
                    list(RETAIL_ACC_TYPES["low"].keys()),
                    p=list(RETAIL_ACC_TYPES["low"].values()),
                )
            elif inc_level == "mid":
                acc_type = random_state.choice(
                    list(RETAIL_ACC_TYPES["mid"].keys()),
                    p=list(RETAIL_ACC_TYPES["mid"].values()),
                )
            else:
                acc_type = random_state.choice(
                    list(RETAIL_ACC_TYPES["high"].keys()),
                    p=list(RETAIL_ACC_TYPES["high"].values()),
                )
            currency = random_state.choice(
                ["AED", "AED", "AED", "USD", "GBP", "EUR"],
                p=[0.7, 0.1, 0.1, 0.05, 0.03, 0.02],
            )
        else:
            acc_type = random_state.choice([
                "Business Current",
                "Business Savings",
                "Trade Finance",
                "Escrow",
                "Payroll",
            ])
            currency = random_state.choice(
                ["AED", "USD", "EUR", "GBP"], p=[0.5, 0.3, 0.1, 0.1]
            )

        acc = {
            "account_id": f"ACC{1000000 + len(accounts)}",
            "customer_id": cust["customer_id"],
            "account_type": acc_type,
            "currency": currency,
            "open_date": cust["account_open_date"],
            "status": "active",
            "current_balance": 0,
            "branch_id": random_state.choice([
                "BR001-DIFC",
                "BR002-Deira",
                "BR003-Marina",
                "BR004-AD-Corniche",
                "BR005-Sharjah",
                "BR006-JAFZA",
            ]),
        }
        accounts.append(acc)

df_accounts = pd.DataFrame(accounts)
print(f"Accounts generated: {len(df_accounts)}")

# =============================================================================
# 4. GENERATE TRANSACTIONS
# =============================================================================

print("Generating transactions (this may take a few minutes)...")

cust_lookup = df_customers.set_index("customer_id")
acc_lookup = df_accounts.set_index("account_id")

all_account_ids = df_accounts["account_id"].tolist()
uae_banks = [
    "Emirates NBD",
    "FAB",
    "ADCB",
    "Mashreq",
    "Dubai Islamic",
    "HSBC UAE",
    "Standard Chartered UAE",
]
exchange_houses = [
    "Al Ansari",
    "UAE Exchange",
    "Lulu Exchange",
    "Wall Street",
    "Al Fardan",
]
offshore_entities = [f"OFFSHORE_{i}" for i in range(500)]
corporate_counterparties = [f"TRADE_PARTNER_{i}" for i in range(1000)]

transactions = []
account_batches = np.array_split(df_accounts, max(1, len(df_accounts) // 500))

for batch_idx, acc_batch in enumerate(
    tqdm(account_batches, desc="Transaction Batches")
):
    batch_txns = []

    for _, acc in acc_batch.iterrows():
        cust = cust_lookup.loc[acc["customer_id"]]
        seg = cust["segment"]
        suspicion = cust["suspicion_label"]
        acc_open = acc["open_date"]
        if isinstance(acc_open, str):
            acc_open = datetime.strptime(acc_open, "%Y-%m-%d")

        period_start = max(acc_open, START_DATE)
        period_end = END_DATE
        if period_start >= period_end:
            continue

        if seg == "retail":
            base_count = random_state.poisson(
                15 if cust["income_level"] == "low" else 25
            )
        elif seg == "sme":
            base_count = random_state.poisson(40)
        elif seg == "corporate":
            base_count = random_state.poisson(80)
        elif seg == "dnfbp":
            base_count = random_state.poisson(35)
        else:
            base_count = random_state.poisson(60)

        n_months = max(1, (period_end - period_start).days // 30)

        for month_offset in range(n_months):
            month_start = period_start + timedelta(days=30 * month_offset)
            month_end = min(month_start + timedelta(days=30), period_end)
            n_txns = max(1, random_state.poisson(base_count))

            for _ in range(n_txns):
                txn_time = random_date(month_start, month_end)

                if seg == "retail":
                    if (
                        cust["employment_type"]
                        in ["Salaried-Private", "Salaried-Govt"]
                        and random_state.random() < 0.25
                    ):
                        txn_type = "Salary Credit"
                    elif random_state.random() < 0.15:
                        txn_type = "Cash Deposit"
                    elif random_state.random() < 0.15:
                        txn_type = "Cash Withdrawal"
                    elif random_state.random() < 0.25:
                        txn_type = "Domestic Transfer"
                    elif random_state.random() < 0.10:
                        txn_type = "International Remittance"
                    else:
                        txn_type = "Card Payment"
                elif seg == "sme":
                    txn_type = random_state.choice(
                        [
                            "Supplier Payment",
                            "Customer Receipt",
                            "Payroll",
                            "Tax Payment",
                            "Loan Repayment",
                            "Cash Deposit",
                            "Transfer",
                        ],
                        p=[0.25, 0.25, 0.15, 0.10, 0.10, 0.08, 0.07],
                    )
                elif seg == "corporate":
                    txn_type = random_state.choice(
                        [
                            "Intercompany Transfer",
                            "Supplier Payment",
                            "Customer Receipt",
                            "Trade Finance",
                            "Payroll",
                            "Treasury",
                            "Dividend",
                        ],
                        p=[0.20, 0.20, 0.20, 0.15, 0.10, 0.10, 0.05],
                    )
                elif seg == "dnfbp":
                    txn_type = random_state.choice(
                        [
                            "Customer Receipt",
                            "Supplier Payment",
                            "Cash Deposit",
                            "Commission",
                            "Transfer",
                            "Investment",
                        ],
                        p=[0.30, 0.20, 0.25, 0.10, 0.10, 0.05],
                    )
                else:
                    txn_type = random_state.choice(
                        [
                            "Fiat Deposit",
                            "Fiat Withdrawal",
                            "Crypto Purchase",
                            "Crypto Sale",
                            "Transfer",
                            "Fee",
                        ],
                        p=[0.25, 0.25, 0.20, 0.20, 0.08, 0.02],
                    )

                if txn_type == "Salary Credit":
                    amount = (
                        int(
                            cust["monthly_income"]
                            * random_state.uniform(0.95, 1.05)
                        )
                        if pd.notna(cust["monthly_income"])
                        else random_state.randint(3000, 15000)
                    )
                elif txn_type in ["Cash Deposit", "Cash Withdrawal"]:
                    amount = int(random_state.lognormal(8, 1.5))
                    amount = max(100, min(500000, amount))
                elif txn_type == "Card Payment":
                    amount = int(random_state.lognormal(5, 1))
                    amount = max(10, min(50000, amount))
                elif txn_type in [
                    "International Remittance",
                    "Fiat Withdrawal",
                ]:
                    amount = int(random_state.lognormal(9, 0.8))
                    amount = max(500, min(100000, amount))
                elif seg in ["sme", "corporate", "dnfbp"]:
                    amount = int(random_state.lognormal(12, 1.5))
                    amount = max(1000, min(5000000, amount))
                else:
                    amount = int(random_state.lognormal(9, 1.2))
                    amount = max(100, min(500000, amount))

                if txn_type in ["Cash Deposit", "Cash Withdrawal"]:
                    channel = random_state.choice(
                        ["Branch", "ATM", "CDM"], p=[0.6, 0.3, 0.1]
                    )
                elif txn_type == "Card Payment":
                    channel = random_state.choice(
                        ["POS", "E-commerce", "Mobile Wallet"],
                        p=[0.6, 0.3, 0.1],
                    )
                else:
                    channel = random_state.choice(
                        ["Internet Banking", "Mobile App", "Branch", "SWIFT", "API"],
                        p=[0.45, 0.35, 0.10, 0.08, 0.02],
                    )

                if txn_type in ["Salary Credit"]:
                    counterparty = cust.get("employer_name", fake.company())
                    cp_bank = random_state.choice(uae_banks)
                    cp_country = "United Arab Emirates"
                elif txn_type == "International Remittance":
                    corridor = random_state.choice(
                        list(CORRIDORS.keys()), p=list(CORRIDORS.values())
                    )
                    counterparty = (
                        f"Beneficiary_{random_state.randint(1, 9999)}"
                    )
                    cp_bank = random_state.choice(
                        exchange_houses + ["Western Union", "MoneyGram"]
                    )
                    cp_country = corridor
                elif txn_type in [
                    "Supplier Payment",
                    "Customer Receipt",
                    "Trade Finance",
                ]:
                    counterparty = random_state.choice(corporate_counterparties)
                    cp_bank = random_state.choice(
                        uae_banks
                        + ["HSBC London", "Citi NY", "Standard Chartered HK"]
                    )
                    cp_country = random_state.choice([
                        "United Arab Emirates",
                        "India",
                        "China",
                        "UK",
                        "USA",
                        "Saudi Arabia",
                        "Offshore",
                    ])
                elif txn_type in [
                    "Intercompany Transfer",
                    "Transfer",
                    "Layering",
                ]:
                    counterparty = random_state.choice(all_account_ids)
                    cp_bank = random_state.choice(uae_banks)
                    cp_country = "United Arab Emirates"
                else:
                    counterparty = random_state.choice(
                        ["Self", "Merchant", "Utility", "Investment", "Loan"]
                        + all_account_ids[:50]
                    )
                    cp_bank = random_state.choice(uae_banks + ["N/A"])
                    cp_country = "United Arab Emirates"

                is_cross_border = cp_country != "United Arab Emirates"
                is_suspicious = 0
                suspicion_type = None

                # Inject suspicious patterns
                if suspicion != "normal":
                    if (
                        suspicion == "structuring"
                        and txn_type == "Cash Deposit"
                        and random_state.random() < 0.7
                    ):
                        amount = random_state.randint(45000, 49500)
                        channel = "Branch"
                        is_suspicious = 1
                        suspicion_type = "structuring"

                    elif (
                        suspicion == "rapid_movement"
                        and amount > 50000
                        and random_state.random() < 0.4
                    ):
                        is_suspicious = 1
                        suspicion_type = "rapid_movement"
                        if random_state.random() < 0.5:
                            txn_type = "Domestic Transfer"
                            cp_country = random_state.choice(
                                ["High-Risk Jurisdiction", "Offshore"]
                            )

                    elif (
                        suspicion == "pep_unusual"
                        and amount > cust.get("monthly_income", 10000) * 5
                    ):
                        is_suspicious = 1
                        suspicion_type = "pep_unusual"
                        txn_type = "Transfer"
                        counterparty = random_state.choice(offshore_entities)
                        cp_country = "Offshore"

                    elif (
                        suspicion == "trade_misinvoicing"
                        and seg in ["sme", "corporate"]
                        and amount > 100000
                    ):
                        if random_state.random() < 0.3:
                            is_suspicious = 1
                            suspicion_type = "trade_misinvoicing"
                            amount = int(amount / 10000) * 10000
                            counterparty = random_state.choice(offshore_entities)
                            cp_country = "Offshore"
                            txn_type = "Supplier Payment"

                    elif (
                        suspicion == "hawala"
                        and txn_type == "International Remittance"
                    ):
                        if random_state.random() < 0.5:
                            is_suspicious = 1
                            suspicion_type = "hawala"
                            amount = random_state.randint(40000, 60000)
                            cp_bank = random_state.choice(exchange_houses)

                txn = {
                    "transaction_id": (
                        f"TXN{900000000 + len(transactions) + len(batch_txns)}"
                    ),
                    "account_id": acc["account_id"],
                    "customer_id": acc["customer_id"],
                    "timestamp": txn_time,
                    "transaction_type": txn_type,
                    "channel": channel,
                    "amount": amount,
                    "currency": acc["currency"],
                    "amount_aed_equivalent": (
                        amount
                        if acc["currency"] == "AED"
                        else int(amount * random_state.uniform(3.65, 3.68))
                    ),
                    "debit_credit": (
                        "credit"
                        if txn_type
                        in [
                            "Salary Credit",
                            "Customer Receipt",
                            "Cash Deposit",
                            "Fiat Deposit",
                        ]
                        else "debit"
                    ),
                    "counterparty_name": counterparty,
                    "counterparty_bank": cp_bank,
                    "counterparty_country": cp_country,
                    "is_cross_border": is_cross_border,
                    "is_cash": txn_type
                    in ["Cash Deposit", "Cash Withdrawal"],
                    "is_suspicious": is_suspicious,
                    "suspicion_type": suspicion_type,
                }
                batch_txns.append(txn)

    transactions.extend(batch_txns)
    if batch_idx % 5 == 0 and batch_idx > 0:
        print(f"  ...processed {len(transactions):,} transactions so far")

df_transactions = pd.DataFrame(transactions)
print(f"Total transactions generated: {len(df_transactions):,}")

# =============================================================================
# 5. POST-PROCESSING & NETWORK FEATURES
# =============================================================================

print("Adding network features...")

retail_cust = df_customers[df_customers["segment"] == "retail"].copy()
n_shared = int(len(retail_cust) * 0.05)
shared_groups = np.array_split(
    random_state.choice(retail_cust.index, size=n_shared, replace=False),
    max(1, n_shared // 5),
)

df_customers["address"] = df_customers.apply(
    lambda x: fake.street_address() + ", " + x["emirate"], axis=1
)
for group in shared_groups:
    shared_addr = (
        fake.street_address() + ", " + retail_cust.loc[group[0], "emirate"]
    )
    df_customers.loc[group, "address"] = shared_addr

n_phone_shared = int(len(retail_cust) * 0.02)
phone_groups = np.array_split(
    random_state.choice(retail_cust.index, size=n_phone_shared, replace=False),
    max(1, n_phone_shared // 3),
)
df_customers["phone_number"] = [
    fake.phone_number() for _ in range(len(df_customers))
]
for group in phone_groups:
    shared_phone = fake.phone_number()
    df_customers.loc[group, "phone_number"] = shared_phone

high_value_idx = df_customers[
    (df_customers["segment"] == "retail")
    & (df_customers["income_level"] == "high")
].index
n_intro = int(len(high_value_idx) * 0.04)
introducers = random_state.choice(
    df_customers[df_customers["segment"] == "retail"]["customer_id"],
    size=n_intro,
)
df_customers["introducer_id"] = None
df_customers.loc[
    random_state.choice(high_value_idx, size=n_intro, replace=False),
    "introducer_id",
] = introducers

sme_idx = df_customers[df_customers["segment"] == "sme"].index
n_beneficial = int(len(sme_idx) * 0.40)
df_customers["beneficial_owner_id"] = None
bo_choices = random_state.choice(
    df_customers[df_customers["segment"] == "retail"]["customer_id"],
    size=n_beneficial,
)
df_customers.loc[
    random_state.choice(sme_idx, size=n_beneficial, replace=False),
    "beneficial_owner_id",
] = bo_choices

# =============================================================================
# 6. VELOCITY FEATURES (for ML model)
# =============================================================================

print("Computing velocity features...")

df_transactions["date"] = pd.to_datetime(df_transactions["timestamp"]).dt.date
daily_vol = (
    df_transactions.groupby(["account_id", "date"])
    .agg(daily_count=("amount", "count"), daily_volume=("amount", "sum"))
    .reset_index()
)

daily_vol = daily_vol.sort_values(["account_id", "date"])
daily_vol["velocity_7d"] = (
    daily_vol.groupby("account_id")["daily_volume"]
    .rolling(7, min_periods=1)
    .sum()
    .reset_index(0, drop=True)
)
daily_vol["velocity_30d"] = (
    daily_vol.groupby("account_id")["daily_volume"]
    .rolling(30, min_periods=1)
    .sum()
    .reset_index(0, drop=True)
)

df_transactions = df_transactions.merge(
    daily_vol[["account_id", "date", "velocity_7d", "velocity_30d"]],
    on=["account_id", "date"],
    how="left",
)

df_transactions = df_transactions.sort_values(
    ["account_id", "transaction_type", "timestamp"]
)
df_transactions["days_since_last_similar"] = (
    df_transactions.groupby(["account_id", "transaction_type"])["timestamp"]
    .diff()
    .dt.days
)

# =============================================================================
# 7. SAVE TO CSV
# =============================================================================

print("Saving datasets...")

df_customers.to_csv(f"{OUTPUT_DIR}/customers.csv", index=False)
df_accounts.to_csv(f"{OUTPUT_DIR}/accounts.csv", index=False)
df_transactions.to_csv(f"{OUTPUT_DIR}/transactions.csv", index=False)

summary = {
    "total_customers": len(df_customers),
    "total_accounts": len(df_accounts),
    "total_transactions": len(df_transactions),
    "suspicious_customers": (
        df_customers["suspicion_label"] != "normal"
    ).sum(),
    "suspicious_transactions": df_transactions["is_suspicious"].sum(),
    "suspicious_breakdown": df_customers[
        df_customers["suspicion_label"] != "normal"
    ]["suspicion_label"]
    .value_counts()
    .to_dict(),
    "segment_breakdown": df_customers["segment"].value_counts().to_dict(),
    "retail_nationality_top5": df_customers[
        df_customers["segment"] == "retail"
    ]["nationality"]
    .value_counts()
    .head(5)
    .to_dict(),
}

summary_df = pd.DataFrame([summary])
summary_df.to_csv(f"{OUTPUT_DIR}/data_summary.csv", index=False)

print("\n" + "=" * 60)
print("DATA GENERATION COMPLETE")
print("=" * 60)
print(f"Output directory: {OUTPUT_DIR}")
print(f"Customers:         {len(df_customers):,}")
print(f"Accounts:          {len(df_accounts):,}")
print(f"Transactions:      {len(df_transactions):,}")
print(
    f"Suspicious customers: {summary['suspicious_customers']:,} ({summary['suspicious_customers']/len(df_customers)*100:.2f}%)"
)
print(
    f"Suspicious transactions: {summary['suspicious_transactions']:,} ({summary['suspicious_transactions']/len(df_transactions)*100:.3f}%)"
)
print("\nSuspicious customer breakdown:")
for k, v in summary["suspicious_breakdown"].items():
    print(f"  {k}: {v}")
print("\nTop 5 retail nationalities:")
for k, v in summary["retail_nationality_top5"].items():
    print(f"  {k}: {v}")
```

```python
# =============================================================================
# UAE AML SYNTHETIC DATA GENERATOR
# For Transaction Monitoring & AML Detection Projects
# =============================================================================

import os
import warnings
from datetime import datetime, timedelta

import numpy as np
import pandas as pd
from faker import Faker

warnings.filterwarnings("ignore")

# --- OPTIONAL PROGRESS BAR ---------------------------------------------------
try:
    from tqdm.notebook import tqdm
except ImportError:

    def tqdm(iterable, **kwargs):
        return iterable


# =============================================================================
# CONFIGURATION
# =============================================================================

SEED = 42
np.random.seed(SEED)
random_state = np.random.RandomState(SEED)
fake = Faker()
fake.seed_instance(SEED)

NUM_CUSTOMERS = 10000  # Scale up as needed
TRANS_MONTHS = 12
START_DATE = datetime(2023, 1, 1)
END_DATE = START_DATE + timedelta(days=30 * TRANS_MONTHS)

OUTPUT_DIR = "./aml_synthetic_uae"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# =============================================================================
# DISTRIBUTIONS & MAPPINGS
# =============================================================================

# --- Customer Segment Mix ----------------------------------------------------
SEGMENTS = {
    "retail": 0.78,
    "sme": 0.12,
    "corporate": 0.06,
    "dnfbp": 0.03,
    "vasp": 0.01,
}

# --- Retail Nationalities (expat-heavy) --------------------------------------
RETAIL_NATS = [
    "India",
    "Pakistan",
    "Philippines",
    "Bangladesh",
    "Egypt",
    "Nepal",
    "Sri Lanka",
    "United Kingdom",
    "Jordan",
    "Sudan",
    "United Arab Emirates",
    "Lebanon",
    "Russia",
    "Nigeria",
    "Iran",
    "China",
    "Other",
]
RETAIL_NAT_WEIGHTS = [
    28,
    13,
    7,
    8,
    6,
    4,
    4,
    4,
    3,
    3,
    16,
    3,
    2,
    1.5,
    1,
    1.5,
    2,
]

# --- Age & Gender ------------------------------------------------------------
AGE_BINS = [18, 25, 35, 45, 55, 65, 80]
AGE_WEIGHTS = [0.12, 0.35, 0.28, 0.18, 0.05, 0.02]
GENDER_WEIGHTS = [0.68, 0.32]  # Male, Female

# --- Emirates ----------------------------------------------------------------
EMIRATES = [
    "Dubai",
    "Abu Dhabi",
    "Sharjah",
    "Ajman",
    "Ras Al Khaimah",
    "Fujairah",
    "Umm Al Quwain",
]
EMIRATE_WEIGHTS = [0.55, 0.25, 0.08, 0.04, 0.03, 0.03, 0.02]

# --- Occupations by Nationality Cluster --------------------------------------
OCCUPATIONS_LOW = [
    "Construction Laborer",
    "Domestic Worker",
    "Delivery Driver",
    "Taxi Driver",
    "Retail Shop Worker",
    "Security Guard",
    "Waiter",
    "Salon Worker",
]
OCCUPATIONS_MID = [
    "Nurse",
    "Teacher",
    "Accountant",
    "Sales Executive",
    "IT Support Technician",
    "Shop Owner",
    "Admin Assistant",
]
OCCUPATIONS_HIGH = [
    "Engineer",
    "Doctor",
    "Business Owner",
    "Finance Manager",
    "IT Consultant",
]

OCCUPATION_INCOME = {
    "Construction Laborer": (1500, 4000),
    "Domestic Worker": (1200, 3500),
    "Delivery Driver": (2000, 5000),
    "Taxi Driver": (2500, 6000),
    "Security Guard": (1800, 4000),
    "Retail Shop Worker": (2000, 5000),
    "Waiter": (2000, 4500),
    "Salon Worker": (2500, 5500),
    "Nurse": (6000, 15000),
    "Teacher": (7000, 18000),
    "Accountant": (8000, 20000),
    "Sales Executive": (7000, 25000),
    "IT Support Technician": (8000, 18000),
    "Shop Owner": (8000, 40000),
    "Engineer": (15000, 45000),
    "Doctor": (25000, 80000),
    "Business Owner": (15000, 100000),
    "Finance Manager": (20000, 60000),
    "IT Consultant": (18000, 50000),
    "Government Official": (15000, 50000),
    "Admin Assistant": (5000, 12000),
}

# --- SME / Corporate / DNFBP / VASP ------------------------------------------
SME_TYPES = [
    "General Trading",
    "Restaurant / F&B",
    "Construction / Contracting",
    "Consultancy",
    "E-commerce",
    "Real Estate Brokerage",
    "Gold / Jewelry Trading",
    "Logistics / Shipping",
    "Freelance",
]
SME_TYPE_WEIGHTS = [0.22, 0.15, 0.12, 0.14, 0.10, 0.08, 0.05, 0.07, 0.07]

CORP_INDUSTRY = [
    "Real Estate Developer",
    "Government-Linked",
    "MNC Regional HQ",
    "Bank / FI",
    "Airline / Logistics",
    "Oil / Energy",
    "Hospitality",
    "Healthcare",
    "Education",
]
CORP_INDUSTRY_WEIGHTS = [
    0.20,
    0.18,
    0.15,
    0.10,
    0.08,
    0.07,
    0.12,
    0.05,
    0.05,
]

DNFBP_TYPES = [
    "Real Estate Broker",
    "Gold Dealer",
    "Luxury Car Dealer",
    "Corporate Service Provider",
    "Lawyer",
    "Accountant",
]
DNFBP_WEIGHTS = [0.35, 0.25, 0.15, 0.15, 0.05, 0.05]

VASP_TYPES = ["Crypto Exchange", "OTC Desk", "Custody Provider", "Broker / ATM"]
VASP_WEIGHTS = [0.40, 0.30, 0.20, 0.10]

# --- Account Types -----------------------------------------------------------
RETAIL_ACC_TYPES = {
    "low": {"Basic Savings": 0.6, "Salary Account": 0.4},
    "mid": {
        "Current Account": 0.5,
        "Basic Savings": 0.3,
        "Fixed Deposit": 0.2,
    },
    "high": {
        "Current Account": 0.4,
        "Wealth / Priority": 0.35,
        "Multi-currency": 0.25,
    },
}

# --- Suspicious Pattern Config -----------------------------------------------
SUSPICIOUS_RATE = 0.035  # 3.5% of customers are "bad actors"
SUSPICION_TYPES = [
    "structuring",
    "rapid_movement",
    "layering",
    "pep_unusual",
    "trade_misinvoicing",
    "hawala",
]

# --- Cross-border corridors ----------------------------------------------------
CORRIDORS = {
    "India": 0.22,
    "Pakistan": 0.12,
    "Philippines": 0.08,
    "UK": 0.07,
    "USA": 0.06,
    "China": 0.05,
    "Saudi Arabia": 0.08,
    "Other GCC": 0.10,
    "Offshore": 0.08,
    "High-Risk": 0.05,
    "Africa": 0.05,
    "Other": 0.04,
}

# =============================================================================
# HELPER FUNCTIONS
# =============================================================================


def random_date(start, end):
    """Return random datetime between start and end."""
    delta = end - start
    return start + timedelta(
        seconds=random_state.randint(0, max(1, int(delta.total_seconds())))
    )


def assign_emirate(nationality, income):
    """Emirate selection based on demographics."""
    if nationality in ["Bangladesh", "Nepal", "Pakistan"] and income < 5000:
        return random_state.choice(
            ["Dubai", "Sharjah", "Ajman"], p=[0.5, 0.3, 0.2]
        )
    elif income > 50000:
        return random_state.choice(["Dubai", "Abu Dhabi"], p=[0.7, 0.3])
    return random_state.choice(EMIRATES, p=EMIRATE_WEIGHTS)


def assign_occupation(nationality, age):
    """Occupation correlated with nationality (Dubai labor market reality)."""
    if nationality in ["Bangladesh", "Nepal"]:
        occs = OCCUPATIONS_LOW
        weights = [0.50, 0.10, 0.15, 0.10, 0.05, 0.05, 0.03, 0.02]
    elif nationality == "Philippines":
        occs = [
            "Domestic Worker",
            "Nurse",
            "Salon Worker",
            "Retail Shop Worker",
            "Teacher",
        ]
        weights = [0.30, 0.25, 0.15, 0.15, 0.15]
    elif nationality in ["India", "Pakistan"]:
        occs = OCCUPATIONS_LOW + OCCUPATIONS_MID + OCCUPATIONS_HIGH
        weights = [
            0.14,
            0.05,
            0.08,
            0.08,
            0.05,
            0.07,
            0.05,
            0.04,
            0.07,
            0.06,
            0.06,
            0.05,
            0.05,
            0.04,
            0.04,
            0.02,
            0.02,
            0.03,
            0.02,
            0.02,
        ]
    elif nationality == "United Arab Emirates":
        occs = [
            "Government Official",
            "Business Owner",
            "Finance Manager",
            "Engineer",
            "Doctor",
            "Teacher",
            "Admin Assistant",
        ]
        weights = [0.30, 0.15, 0.15, 0.15, 0.10, 0.10, 0.05]
    elif nationality in ["United Kingdom", "Russia"]:
        occs = [
            "Finance Manager",
            "IT Consultant",
            "Business Owner",
            "Engineer",
            "Doctor",
        ]
        weights = [0.25, 0.20, 0.20, 0.20, 0.15]
    elif nationality == "Egypt":
        occs = [
            "Accountant",
            "Teacher",
            "Sales Executive",
            "Engineer",
            "Nurse",
            "Admin Assistant",
        ]
        weights = [0.20, 0.20, 0.20, 0.20, 0.10, 0.10]
    else:
        occs = OCCUPATIONS_MID + OCCUPATIONS_HIGH
        weights = [
            0.10,
            0.10,
            0.10,
            0.10,
            0.10,
            0.10,
            0.08,
            0.07,
            0.07,
            0.06,
            0.06,
            0.06,
        ]

    return random_state.choice(occs, p=np.array(weights) / sum(weights))


def income_from_occupation(occ):
    lo, hi = OCCUPATION_INCOME.get(occ, (5000, 15000))
    return int(random_state.lognormal(np.log((lo + hi) / 2), 0.3))


def kyc_risk_rating(nationality, occupation, income, pep_flag, is_sme=False):
    score = 0
    if pep_flag:
        score += 40
    if nationality in ["Iran", "Syria", "North Korea", "Afghanistan"]:
        score += 35
    if occupation in ["Business Owner", "Shop Owner", "Gold Dealer"]:
        score += 15
    if income > 80000:
        score += 10
    if is_sme:
        score += 10
    if nationality in ["Nigeria", "Russia"]:
        score += 10

    if score >= 40:
        return "High"
    if score >= 20:
        return "Medium"
    return "Low"


# =============================================================================
# 1. GENERATE CUSTOMERS
# =============================================================================

print("Generating customers...")
customers = []

seg_counts = {k: int(NUM_CUSTOMERS * v) for k, v in SEGMENTS.items()}
seg_counts["retail"] += NUM_CUSTOMERS - sum(seg_counts.values())

# --- RETAIL ---
for i in tqdm(range(seg_counts["retail"]), desc="Retail"):
    nat = random_state.choice(
        RETAIL_NATS, p=np.array(RETAIL_NAT_WEIGHTS) / sum(RETAIL_NAT_WEIGHTS)
    )
    age = int(random_state.choice([22, 30, 40, 50, 60, 70], p=AGE_WEIGHTS))
    gender = random_state.choice(["M", "F"], p=GENDER_WEIGHTS)

    if nat == "United Arab Emirates" and random_state.random() < 0.4:
        gender = "F"
    if nat == "Philippines" and random_state.random() < 0.6:
        gender = "F"

    occ = assign_occupation(nat, age)
    income = income_from_occupation(occ)

    pep_flag = random_state.random() < 0.018
    if pep_flag and nat != "United Arab Emirates":
        pep_flag = random_state.random() < 0.6

    cust = {
        "customer_id": f"CUST{100000 + i}",
        "segment": "retail",
        "customer_type": "individual",
        "name": fake.name_male() if gender == "M" else fake.name_female(),
        "nationality": nat,
        "gender": gender,
        "age": age,
        "occupation": occ,
        "monthly_income": income,
        "income_level": (
            "low"
            if income < 8000
            else ("mid" if income < 25000 else "high")
        ),
        "pep_flag": pep_flag,
        "pep_category": (
            random_state.choice(["domestic", "foreign", "international_org"])
            if pep_flag
            else None
        ),
        "sanctions_flag": random_state.random() < 0.0005,
        "adverse_media_flag": random_state.random() < 0.003,
        "high_risk_jurisdiction_flag": nat
        in ["Iran", "Syria", "Afghanistan", "North Korea"],
        "kyc_risk_rating": kyc_risk_rating(nat, occ, income, pep_flag),
        "account_open_date": random_date(
            START_DATE - timedelta(days=365 * 3),
            END_DATE - timedelta(days=180),
        ),
        "emirate": assign_emirate(nat, income),
        "visa_type": random_state.choice(
            ["Employment", "Investor", "Tourist", "Golden", "Student", "None"]
        ),
        "employment_type": random_state.choice(
            [
                "Salaried-Private",
                "Salaried-Govt",
                "Self-employed",
                "Business Owner",
                "Unemployed",
                "Student",
                "Retired",
            ],
            p=[0.55, 0.12, 0.15, 0.10, 0.03, 0.03, 0.02],
        ),
        "suspicion_label": "normal",
        "shell_company_risk_score": 0,
    }
    customers.append(cust)

# --- SME / FREELANCER ---
for i in tqdm(range(seg_counts["sme"]), desc="SME"):
    sme_type = random_state.choice(SME_TYPES, p=SME_TYPE_WEIGHTS)
    turnover = random_state.choice([
        random_state.lognormal(13, 1),
        random_state.lognormal(16, 0.5),
        random_state.lognormal(18, 0.8),
    ])
    turnover = max(50000, min(50000000, turnover))

    cust = {
        "customer_id": f"SME{200000 + i}",
        "segment": "sme",
        "customer_type": "entity",
        "name": fake.company(),
        "nationality": random_state.choice(
            ["India", "Pakistan", "UAE", "UK", "Lebanon", "China", "Russia", "Other"],
            p=[0.25, 0.10, 0.20, 0.10, 0.10, 0.10, 0.05, 0.10],
        ),
        "gender": None,
        "age": None,
        "occupation": sme_type,
        "monthly_income": None,
        "annual_turnover": int(turnover),
        "income_level": "business",
        "pep_flag": random_state.random() < 0.03,
        "pep_category": None,
        "sanctions_flag": random_state.random() < 0.001,
        "adverse_media_flag": random_state.random() < 0.005,
        "high_risk_jurisdiction_flag": False,
        "kyc_risk_rating": kyc_risk_rating(
            "Other", sme_type, turnover, False, is_sme=True
        ),
        "account_open_date": random_date(
            START_DATE - timedelta(days=365 * 2), END_DATE - timedelta(days=90)
        ),
        "emirate": random_state.choice(EMIRATES, p=EMIRATE_WEIGHTS),
        "visa_type": None,
        "employment_type": None,
        "suspicion_label": "normal",
        "shell_company_risk_score": (
            random_state.randint(0, 100)
            if random_state.random() < 0.03
            else random_state.randint(0, 30)
        ),
        "business_size_employees": random_state.choice(
            [1, 3, 10, 30, 100, 250], p=[0.20, 0.35, 0.30, 0.11, 0.03, 0.01]
        ),
        "legal_structure": random_state.choice(
            ["Sole Establishment", "LLC", "Free Zone", "Branch", "Civil"],
            p=[0.40, 0.25, 0.25, 0.07, 0.03],
        ),
    }
    customers.append(cust)

# --- CORPORATE ---
for i in tqdm(range(seg_counts["corporate"]), desc="Corporate"):
    industry = random_state.choice(CORP_INDUSTRY, p=CORP_INDUSTRY_WEIGHTS)
    turnover = random_state.lognormal(20, 1.5)
    turnover = max(1000000, min(500000000, turnover))

    cust = {
        "customer_id": f"CORP{300000 + i}",
        "segment": "corporate",
        "customer_type": "entity",
        "name": fake.company()
        + " "
        + random_state.choice(["Group", "LLC", "Holding", "Ltd", "FZE"]),
        "nationality": "United Arab Emirates",
        "gender": None,
        "age": None,
        "occupation": industry,
        "monthly_income": None,
        "annual_turnover": int(turnover),
        "income_level": "corporate",
        "pep_flag": random_state.random() < 0.08,
        "pep_category": (
            "domestic" if random_state.random() < 0.8 else "foreign"
        ),
        "sanctions_flag": False,
        "adverse_media_flag": random_state.random() < 0.01,
        "high_risk_jurisdiction_flag": False,
        "kyc_risk_rating": random_state.choice(
            ["Low", "Medium"], p=[0.7, 0.3]
        ),
        "account_open_date": random_date(
            START_DATE - timedelta(days=365 * 5),
            END_DATE - timedelta(days=365),
        ),
        "emirate": random_state.choice(["Dubai", "Abu Dhabi"], p=[0.6, 0.4]),
        "visa_type": None,
        "employment_type": None,
        "suspicion_label": "normal",
        "shell_company_risk_score": random_state.randint(0, 15),
    }
    customers.append(cust)

# --- DNFBP ---
for i in tqdm(range(seg_counts["dnfbp"]), desc="DNFBP"):
    dnfbp_type = random_state.choice(DNFBP_TYPES, p=DNFBP_WEIGHTS)
    cust = {
        "customer_id": f"DNFBP{400000 + i}",
        "segment": "dnfbp",
        "customer_type": "entity",
        "name": fake.company() + " " + dnfbp_type.split()[-1],
        "nationality": random_state.choice(
            ["UAE", "Lebanon", "India", "Iran", "UK", "Other"],
            p=[0.40, 0.20, 0.15, 0.10, 0.10, 0.05],
        ),
        "gender": None,
        "age": None,
        "occupation": dnfbp_type,
        "monthly_income": None,
        "annual_turnover": int(random_state.lognormal(16, 1.2)),
        "income_level": "business",
        "pep_flag": random_state.random() < 0.05,
        "pep_category": None,
        "sanctions_flag": random_state.random() < 0.002,
        "adverse_media_flag": random_state.random() < 0.02,
        "high_risk_jurisdiction_flag": False,
        "kyc_risk_rating": (
            "High"
            if dnfbp_type == "Gold Dealer"
            else random_state.choice(["Medium", "High"], p=[0.4, 0.6])
        ),
        "account_open_date": random_date(
            START_DATE - timedelta(days=365 * 3),
            END_DATE - timedelta(days=180),
        ),
        "emirate": "Dubai",
        "visa_type": None,
        "employment_type": None,
        "suspicion_label": "normal",
        "shell_company_risk_score": (
            random_state.randint(10, 60)
            if dnfbp_type == "Corporate Service Provider"
            else random_state.randint(0, 40)
        ),
    }
    customers.append(cust)

# --- VASP ---
for i in tqdm(range(seg_counts["vasp"]), desc="VASP"):
    vasp_type = random_state.choice(VASP_TYPES, p=VASP_WEIGHTS)
    cust = {
        "customer_id": f"VASP{500000 + i}",
        "segment": "vasp",
        "customer_type": "entity",
        "name": fake.company() + " Digital Assets",
        "nationality": random_state.choice(
            ["UAE", "USA", "UK", "Singapore", "Other"],
            p=[0.50, 0.15, 0.15, 0.10, 0.10],
        ),
        "gender": None,
        "age": None,
        "occupation": vasp_type,
        "monthly_income": None,
        "annual_turnover": int(random_state.lognormal(17, 1.5)),
        "income_level": "business",
        "pep_flag": False,
        "pep_category": None,
        "sanctions_flag": False,
        "adverse_media_flag": random_state.random() < 0.01,
        "high_risk_jurisdiction_flag": False,
        "kyc_risk_rating": "High",
        "account_open_date": random_date(
            START_DATE - timedelta(days=365 * 2), END_DATE - timedelta(days=90)
        ),
        "emirate": "Dubai",
        "visa_type": None,
        "employment_type": None,
        "suspicion_label": "normal",
        "shell_company_risk_score": random_state.randint(0, 30),
        "license_status": random_state.choice(
            ["Licensed", "Unlicensed"], p=[0.85, 0.15]
        ),
    }
    customers.append(cust)

df_customers = pd.DataFrame(customers)
print(f"Customers generated: {len(df_customers)}")

# =============================================================================
# 2. ASSIGN SUSPICIOUS LABELS TO SUBSET OF CUSTOMERS
# =============================================================================

print("Injecting suspicious actor labels...")
n_suspicious = int(len(df_customers) * SUSPICIOUS_RATE)
suspicious_idx = random_state.choice(
    df_customers.index, size=n_suspicious, replace=False
)

suspicion_draw = random_state.choice(
    SUSPICION_TYPES,
    size=n_suspicious,
    p=[0.25, 0.20, 0.20, 0.15, 0.12, 0.08],
)

for idx, stype in zip(suspicious_idx, suspicion_draw):
    seg = df_customers.loc[idx, "segment"]
    if seg == "retail" and stype in ["trade_misinvoicing"]:
        stype = random_state.choice(
            ["structuring", "rapid_movement", "layering", "hawala"]
        )
    if seg in ["corporate", "vasp"] and stype in ["structuring", "hawala"]:
        stype = random_state.choice(
            ["rapid_movement", "layering", "pep_unusual", "trade_misinvoicing"]
        )

    df_customers.at[idx, "suspicion_label"] = stype
    df_customers.at[idx, "kyc_risk_rating"] = "High"

print(
    f"Suspicious customers: {(df_customers['suspicion_label'] != 'normal').sum()}"
)

# =============================================================================
# 3. GENERATE ACCOUNTS
# =============================================================================

print("Generating accounts...")
accounts = []

for _, cust in tqdm(
    df_customers.iterrows(), total=len(df_customers), desc="Accounts"
):
    n_accounts = random_state.choice([1, 2, 3], p=[0.65, 0.25, 0.10])
    if cust["segment"] in ["sme", "corporate", "dnfbp"]:
        n_accounts = random_state.choice(
            [1, 2, 3, 4], p=[0.30, 0.30, 0.25, 0.15]
        )

    for acc_i in range(n_accounts):
        if cust["segment"] == "retail":
            inc_level = cust["income_level"]
            if inc_level == "low":
                acc_type = random_state.choice(
                    list(RETAIL_ACC_TYPES["low"].keys()),
                    p=list(RETAIL_ACC_TYPES["low"].values()),
                )
            elif inc_level == "mid":
                acc_type = random_state.choice(
                    list(RETAIL_ACC_TYPES["mid"].keys()),
                    p=list(RETAIL_ACC_TYPES["mid"].values()),
                )
            else:
                acc_type = random_state.choice(
                    list(RETAIL_ACC_TYPES["high"].keys()),
                    p=list(RETAIL_ACC_TYPES["high"].values()),
                )
            currency = random_state.choice(
                ["AED", "AED", "AED", "USD", "GBP", "EUR"],
                p=[0.7, 0.1, 0.1, 0.05, 0.03, 0.02],
            )
        else:
            acc_type = random_state.choice([
                "Business Current",
                "Business Savings",
                "Trade Finance",
                "Escrow",
                "Payroll",
            ])
            currency = random_state.choice(
                ["AED", "USD", "EUR", "GBP"], p=[0.5, 0.3, 0.1, 0.1]
            )

        acc = {
            "account_id": f"ACC{1000000 + len(accounts)}",
            "customer_id": cust["customer_id"],
            "account_type": acc_type,
            "currency": currency,
            "open_date": cust["account_open_date"],
            "status": "active",
            "current_balance": 0,
            "branch_id": random_state.choice([
                "BR001-DIFC",
                "BR002-Deira",
                "BR003-Marina",
                "BR004-AD-Corniche",
                "BR005-Sharjah",
                "BR006-JAFZA",
            ]),
        }
        accounts.append(acc)

df_accounts = pd.DataFrame(accounts)
print(f"Accounts generated: {len(df_accounts)}")

# =============================================================================
# 4. GENERATE TRANSACTIONS
# =============================================================================

print("Generating transactions (this may take a few minutes)...")

cust_lookup = df_customers.set_index("customer_id")
acc_lookup = df_accounts.set_index("account_id")

all_account_ids = df_accounts["account_id"].tolist()
uae_banks = [
    "Emirates NBD",
    "FAB",
    "ADCB",
    "Mashreq",
    "Dubai Islamic",
    "HSBC UAE",
    "Standard Chartered UAE",
]
exchange_houses = [
    "Al Ansari",
    "UAE Exchange",
    "Lulu Exchange",
    "Wall Street",
    "Al Fardan",
]
offshore_entities = [f"OFFSHORE_{i}" for i in range(500)]
corporate_counterparties = [f"TRADE_PARTNER_{i}" for i in range(1000)]

transactions = []

# Split using native DataFrame slicing to keep slices as DataFrames
batch_size = 500
account_batches = [
    df_accounts.iloc[i : i + batch_size]
    for i in range(0, len(df_accounts), batch_size)
]

for batch_idx, acc_batch in enumerate(
    tqdm(account_batches, desc="Transaction Batches")
):
    batch_txns = []

    for _, acc in acc_batch.iterrows():
        cust = cust_lookup.loc[acc["customer_id"]]
        seg = cust["segment"]
        suspicion = cust["suspicion_label"]
        acc_open = acc["open_date"]
        if isinstance(acc_open, str):
            acc_open = datetime.strptime(acc_open, "%Y-%m-%d")

        period_start = max(acc_open, START_DATE)
        period_end = END_DATE
        if period_start >= period_end:
            continue

        if seg == "retail":
            base_count = random_state.poisson(
                15 if cust["income_level"] == "low" else 25
            )
        elif seg == "sme":
            base_count = random_state.poisson(40)
        elif seg == "corporate":
            base_count = random_state.poisson(80)
        elif seg == "dnfbp":
            base_count = random_state.poisson(35)
        else:
            base_count = random_state.poisson(60)

        n_months = max(1, (period_end - period_start).days // 30)

        for month_offset in range(n_months):
            month_start = period_start + timedelta(days=30 * month_offset)
            month_end = min(month_start + timedelta(days=30), period_end)
            n_txns = max(1, random_state.poisson(base_count))

            for _ in range(n_txns):
                txn_time = random_date(month_start, month_end)

                if seg == "retail":
                    if (
                        cust["employment_type"]
                        in ["Salaried-Private", "Salaried-Govt"]
                        and random_state.random() < 0.25
                    ):
                        txn_type = "Salary Credit"
                    elif random_state.random() < 0.15:
                        txn_type = "Cash Deposit"
                    elif random_state.random() < 0.15:
                        txn_type = "Cash Withdrawal"
                    elif random_state.random() < 0.25:
                        txn_type = "Domestic Transfer"
                    elif random_state.random() < 0.10:
                        txn_type = "International Remittance"
                    else:
                        txn_type = "Card Payment"
                elif seg == "sme":
                    txn_type = random_state.choice(
                        [
                            "Supplier Payment",
                            "Customer Receipt",
                            "Payroll",
                            "Tax Payment",
                            "Loan Repayment",
                            "Cash Deposit",
                            "Transfer",
                        ],
                        p=[0.25, 0.25, 0.15, 0.10, 0.10, 0.08, 0.07],
                    )
                elif seg == "corporate":
                    txn_type = random_state.choice(
                        [
                            "Intercompany Transfer",
                            "Supplier Payment",
                            "Customer Receipt",
                            "Trade Finance",
                            "Payroll",
                            "Treasury",
                            "Dividend",
                        ],
                        p=[0.20, 0.20, 0.20, 0.15, 0.10, 0.10, 0.05],
                    )
                elif seg == "dnfbp":
                    txn_type = random_state.choice(
                        [
                            "Customer Receipt",
                            "Supplier Payment",
                            "Cash Deposit",
                            "Commission",
                            "Transfer",
                            "Investment",
                        ],
                        p=[0.30, 0.20, 0.25, 0.10, 0.10, 0.05],
                    )
                else:
                    txn_type = random_state.choice(
                        [
                            "Fiat Deposit",
                            "Fiat Withdrawal",
                            "Crypto Purchase",
                            "Crypto Sale",
                            "Transfer",
                            "Fee",
                        ],
                        p=[0.25, 0.25, 0.20, 0.20, 0.08, 0.02],
                    )

                if txn_type == "Salary Credit":
                    amount = (
                        int(
                            cust["monthly_income"]
                            * random_state.uniform(0.95, 1.05)
                        )
                        if pd.notna(cust["monthly_income"])
                        else random_state.randint(3000, 15000)
                    )
                elif txn_type in ["Cash Deposit", "Cash Withdrawal"]:
                    amount = int(random_state.lognormal(8, 1.5))
                    amount = max(100, min(500000, amount))
                elif txn_type == "Card Payment":
                    amount = int(random_state.lognormal(5, 1))
                    amount = max(10, min(50000, amount))
                elif txn_type in [
                    "International Remittance",
                    "Fiat Withdrawal",
                ]:
                    amount = int(random_state.lognormal(9, 0.8))
                    amount = max(500, min(100000, amount))
                elif seg in ["sme", "corporate", "dnfbp"]:
                    amount = int(random_state.lognormal(12, 1.5))
                    amount = max(1000, min(5000000, amount))
                else:
                    amount = int(random_state.lognormal(9, 1.2))
                    amount = max(100, min(500000, amount))

                if txn_type in ["Cash Deposit", "Cash Withdrawal"]:
                    channel = random_state.choice(
                        ["Branch", "ATM", "CDM"], p=[0.6, 0.3, 0.1]
                    )
                elif txn_type == "Card Payment":
                    channel = random_state.choice(
                        ["POS", "E-commerce", "Mobile Wallet"],
                        p=[0.6, 0.3, 0.1],
                    )
                else:
                    channel = random_state.choice(
                        ["Internet Banking", "Mobile App", "Branch", "SWIFT", "API"],
                        p=[0.45, 0.35, 0.10, 0.08, 0.02],
                    )

                if txn_type in ["Salary Credit"]:
                    counterparty = cust.get("employer_name", fake.company())
                    cp_bank = random_state.choice(uae_banks)
                    cp_country = "United Arab Emirates"
                elif txn_type == "International Remittance":
                    corridor = random_state.choice(
                        list(CORRIDORS.keys()), p=list(CORRIDORS.values())
                    )
                    counterparty = (
                        f"Beneficiary_{random_state.randint(1, 9999)}"
                    )
                    cp_bank = random_state.choice(
                        exchange_houses + ["Western Union", "MoneyGram"]
                    )
                    cp_country = corridor
                elif txn_type in [
                    "Supplier Payment",
                    "Customer Receipt",
                    "Trade Finance",
                ]:
                    counterparty = random_state.choice(corporate_counterparties)
                    cp_bank = random_state.choice(
                        uae_banks
                        + ["HSBC London", "Citi NY", "Standard Chartered HK"]
                    )
                    cp_country = random_state.choice([
                        "United Arab Emirates",
                        "India",
                        "China",
                        "UK",
                        "USA",
                        "Saudi Arabia",
                        "Offshore",
                    ])
                elif txn_type in [
                    "Intercompany Transfer",
                    "Transfer",
                    "Layering",
                ]:
                    counterparty = random_state.choice(all_account_ids)
                    cp_bank = random_state.choice(uae_banks)
                    cp_country = "United Arab Emirates"
                else:
                    counterparty = random_state.choice(
                        ["Self", "Merchant", "Utility", "Investment", "Loan"]
                        + all_account_ids[:50]
                    )
                    cp_bank = random_state.choice(uae_banks + ["N/A"])
                    cp_country = "United Arab Emirates"

                is_cross_border = cp_country != "United Arab Emirates"
                is_suspicious = 0
                suspicion_type = None

                # Inject suspicious patterns
                if suspicion != "normal":
                    if (
                        suspicion == "structuring"
                        and txn_type == "Cash Deposit"
                        and random_state.random() < 0.7
                    ):
                        amount = random_state.randint(45000, 49500)
                        channel = "Branch"
                        is_suspicious = 1
                        suspicion_type = "structuring"

                    elif (
                        suspicion == "rapid_movement"
                        and amount > 50000
                        and random_state.random() < 0.4
                    ):
                        is_suspicious = 1
                        suspicion_type = "rapid_movement"
                        if random_state.random() < 0.5:
                            txn_type = "Domestic Transfer"
                            cp_country = random_state.choice(
                                ["High-Risk Jurisdiction", "Offshore"]
                            )

                    elif (
                        suspicion == "pep_unusual"
                        and amount > cust.get("monthly_income", 10000) * 5
                    ):
                        is_suspicious = 1
                        suspicion_type = "pep_unusual"
                        txn_type = "Transfer"
                        counterparty = random_state.choice(offshore_entities)
                        cp_country = "Offshore"

                    elif (
                        suspicion == "trade_misinvoicing"
                        and seg in ["sme", "corporate"]
                        and amount > 100000
                    ):
                        if random_state.random() < 0.3:
                            is_suspicious = 1
                            suspicion_type = "trade_misinvoicing"
                            amount = int(amount / 10000) * 10000
                            counterparty = random_state.choice(offshore_entities)
                            cp_country = "Offshore"
                            txn_type = "Supplier Payment"

                    elif (
                        suspicion == "hawala"
                        and txn_type == "International Remittance"
                    ):
                        if random_state.random() < 0.5:
                            is_suspicious = 1
                            suspicion_type = "hawala"
                            amount = random_state.randint(40000, 60000)
                            cp_bank = random_state.choice(exchange_houses)

                txn = {
                    "transaction_id": (
                        f"TXN{900000000 + len(transactions) + len(batch_txns)}"
                    ),
                    "account_id": acc["account_id"],
                    "customer_id": acc["customer_id"],
                    "timestamp": txn_time,
                    "transaction_type": txn_type,
                    "channel": channel,
                    "amount": amount,
                    "currency": acc["currency"],
                    "amount_aed_equivalent": (
                        amount
                        if acc["currency"] == "AED"
                        else int(amount * random_state.uniform(3.65, 3.68))
                    ),
                    "debit_credit": (
                        "credit"
                        if txn_type
                        in [
                            "Salary Credit",
                            "Customer Receipt",
                            "Cash Deposit",
                            "Fiat Deposit",
                        ]
                        else "debit"
                    ),
                    "counterparty_name": counterparty,
                    "counterparty_bank": cp_bank,
                    "counterparty_country": cp_country,
                    "is_cross_border": is_cross_border,
                    "is_cash": txn_type
                    in ["Cash Deposit", "Cash Withdrawal"],
                    "is_suspicious": is_suspicious,
                    "suspicion_type": suspicion_type,
                }
                batch_txns.append(txn)

    transactions.extend(batch_txns)
    if batch_idx % 5 == 0 and batch_idx > 0:
        print(f"  ...processed {len(transactions):,} transactions so far")

df_transactions = pd.DataFrame(transactions)
print(f"Total transactions generated: {len(df_transactions):,}")

# =============================================================================
# 5. POST-PROCESSING & NETWORK FEATURES
# =============================================================================

print("Adding network features...")

retail_cust = df_customers[df_customers["segment"] == "retail"].copy()
n_shared = int(len(retail_cust) * 0.05)
shared_groups = np.array_split(
    random_state.choice(retail_cust.index, size=n_shared, replace=False),
    max(1, n_shared // 5),
)

df_customers["address"] = df_customers.apply(
    lambda x: fake.street_address() + ", " + x["emirate"], axis=1
)
for group in shared_groups:
    shared_addr = (
        fake.street_address() + ", " + retail_cust.loc[group[0], "emirate"]
    )
    df_customers.loc[group, "address"] = shared_addr

n_phone_shared = int(len(retail_cust) * 0.02)
phone_groups = np.array_split(
    random_state.choice(retail_cust.index, size=n_phone_shared, replace=False),
    max(1, n_phone_shared // 3),
)
df_customers["phone_number"] = [
    fake.phone_number() for _ in range(len(df_customers))
]
for group in phone_groups:
    shared_phone = fake.phone_number()
    df_customers.loc[group, "phone_number"] = shared_phone

high_value_idx = df_customers[
    (df_customers["segment"] == "retail")
    & (df_customers["income_level"] == "high")
].index
n_intro = int(len(high_value_idx) * 0.04)
introducers = random_state.choice(
    df_customers[df_customers["segment"] == "retail"]["customer_id"],
    size=n_intro,
)
df_customers["introducer_id"] = None
df_customers.loc[
    random_state.choice(high_value_idx, size=n_intro, replace=False),
    "introducer_id",
] = introducers

sme_idx = df_customers[df_customers["segment"] == "sme"].index
n_beneficial = int(len(sme_idx) * 0.40)
df_customers["beneficial_owner_id"] = None
bo_choices = random_state.choice(
    df_customers[df_customers["segment"] == "retail"]["customer_id"],
    size=n_beneficial,
)
df_customers.loc[
    random_state.choice(sme_idx, size=n_beneficial, replace=False),
    "beneficial_owner_id",
] = bo_choices

# =============================================================================
# 6. VELOCITY FEATURES (for ML model)
# =============================================================================

print("Computing velocity features...")

df_transactions["date"] = pd.to_datetime(df_transactions["timestamp"]).dt.date
daily_vol = (
    df_transactions.groupby(["account_id", "date"])
    .agg(daily_count=("amount", "count"), daily_volume=("amount", "sum"))
    .reset_index()
)

daily_vol = daily_vol.sort_values(["account_id", "date"])
daily_vol["velocity_7d"] = (
    daily_vol.groupby("account_id")["daily_volume"]
    .rolling(7, min_periods=1)
    .sum()
    .reset_index(level=0, drop=True)
)
daily_vol["velocity_30d"] = (
    daily_vol.groupby("account_id")["daily_volume"]
    .rolling(30, min_periods=1)
    .sum()
    .reset_index(level=0, drop=True)
)

df_transactions = df_transactions.merge(
    daily_vol[["account_id", "date", "velocity_7d", "velocity_30d"]],
    on=["account_id", "date"],
    how="left",
)

df_transactions = df_transactions.sort_values(
    ["account_id", "transaction_type", "timestamp"]
)
df_transactions["days_since_last_similar"] = (
    df_transactions.groupby(["account_id", "transaction_type"])["timestamp"]
    .diff()
    .dt.days
)

# =============================================================================
# 7. SAVE TO CSV
# =============================================================================

print("Saving datasets...")

df_customers.to_csv(f"{OUTPUT_DIR}/customers.csv", index=False)
df_accounts.to_csv(f"{OUTPUT_DIR}/accounts.csv", index=False)
df_transactions.to_csv(f"{OUTPUT_DIR}/transactions.csv", index=False)

summary = {
    "total_customers": len(df_customers),
    "total_accounts": len(df_accounts),
    "total_transactions": len(df_transactions),
    "suspicious_customers": (
        df_customers["suspicion_label"] != "normal"
    ).sum(),
    "suspicious_transactions": df_transactions["is_suspicious"].sum(),
    "suspicious_breakdown": df_customers[
        df_customers["suspicion_label"] != "normal"
    ]["suspicion_label"]
    .value_counts()
    .to_dict(),
    "segment_breakdown": df_customers["segment"].value_counts().to_dict(),
    "retail_nationality_top5": df_customers[
        df_customers["segment"] == "retail"
    ]["nationality"]
    .value_counts()
    .head(5)
    .to_dict(),
}

summary_df = pd.DataFrame([summary])
summary_df.to_csv(f"{OUTPUT_DIR}/data_summary.csv", index=False)

print("\n" + "=" * 60)
print("DATA GENERATION COMPLETE")
print("=" * 60)
print(f"Output directory: {OUTPUT_DIR}")
print(f"Customers:         {len(df_customers):,}")
print(f"Accounts:          {len(df_accounts):,}")
print(f"Transactions:      {len(df_transactions):,}")
print(
    f"Suspicious customers: {summary['suspicious_customers']:,} ({summary['suspicious_customers']/len(df_customers)*100:.2f}%)"
)
print(
    f"Suspicious transactions: {summary['suspicious_transactions']:,} ({summary['suspicious_transactions']/len(df_transactions)*100:.3f}%)"
)
print("\nSuspicious customer breakdown:")
for k, v in summary["suspicious_breakdown"].items():
    print(f"  {k}: {v}")
print("\nTop 5 retail nationalities:")
for k, v in summary["retail_nationality_top5"].items():
    print(f"  {k}: {v}")
```

```python
# =============================================================================
# UAE AML SYNTHETIC DATA GENERATOR - ADVANCED BEHAVIORAL EDITION (FIXED)
# =============================================================================

import os
import csv
import uuid
import warnings
from datetime import datetime, timedelta

import numpy as np
import pandas as pd
from faker import Faker

warnings.filterwarnings("ignore")

try:
    from tqdm.notebook import tqdm
except ImportError:
    def tqdm(iterable, **kwargs):
        return iterable

# =============================================================================
# 0. CONFIGURATION & SETUP
# =============================================================================

SEED = 42
np.random.seed(SEED)
random_state = np.random.RandomState(SEED)
fake = Faker()
fake.seed_instance(SEED)

# Generation scale
NUM_CUSTOMERS = 10000  
TRANS_MONTHS = 12
START_DATE = pd.Timestamp("2023-01-01")
END_DATE = START_DATE + pd.DateOffset(months=TRANS_MONTHS)

OUTPUT_DIR = "./aml_synthetic_advanced"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# FX Rates (1 unit of currency = X AED)
FX_RATES = {'AED': 1.0, 'USD': 3.6725, 'EUR': 3.95, 'GBP': 4.60, 'INR': 0.044, 'SAR': 0.98}
CURRENCIES = list(FX_RATES.keys())

# =============================================================================
# 1. MASTER ENTITY GENERATORS
# =============================================================================
print("Generating Master Entity Tables...")

# 1A. MERCHANTS
MERCHANT_CATEGORIES = ['Supermarket', 'Telecom', 'Utility', 'Rent', 'Restaurant', 'Fuel', 'Healthcare', 'E-commerce', 'Luxury']
merchants = []
for i in range(500):
    cat = random_state.choice(MERCHANT_CATEGORIES, p=[0.25, 0.1, 0.1, 0.05, 0.2, 0.1, 0.05, 0.1, 0.05])
    merchants.append({
        'merchant_id': f"MERCH_{i+1000}",
        'merchant_name': fake.company() + f" {cat}",
        'category': cat,
        'country': 'United Arab Emirates' if cat not in ['E-commerce'] else random_state.choice(['United Arab Emirates', 'USA', 'UK'])
    })
df_merchants = pd.DataFrame(merchants)

# 1B. EMPLOYERS
employers = []
for i in range(1000):
    employers.append({
        'employer_id': f"EMP_{i+1000}",
        'employer_name': fake.company() + " LLC",
        'industry': random_state.choice(['Construction', 'Tech', 'Healthcare', 'Finance', 'Education', 'Retail'])
    })
df_employers = pd.DataFrame(employers)

# 1C. EXTERNAL COUNTERPARTIES (For transfers)
counterparties = []
for i in range(2000):
    cp_type = random_state.choice(['Individual', 'Corporate', 'Offshore', 'High-Risk'])
    if cp_type == 'Individual':
        country = random_state.choice(['India', 'Pakistan', 'Philippines', 'UK', 'Egypt'])
        name = fake.name()
    elif cp_type == 'Offshore':
        country = random_state.choice(['Cayman Islands', 'BVI', 'Panama', 'Seychelles'])
        name = fake.company() + " Holdings"
    elif cp_type == 'High-Risk':
        country = random_state.choice(['Iran', 'Syria', 'North Korea', 'Myanmar'])
        name = fake.company() + " Trading"
    else:
        country = random_state.choice(['USA', 'Saudi Arabia', 'China', 'Germany'])
        name = fake.company() + " Corp"
        
    counterparties.append({
        'cp_id': f"CP_{i+10000}",
        'cp_name': name,
        'cp_type': cp_type,
        'country': country,
        'bank_name': fake.company() + " Bank"
    })
df_counterparties = pd.DataFrame(counterparties)

# =============================================================================
# 2. CUSTOMER & BEHAVIORAL PROFILE GENERATION
# =============================================================================
print("Generating Customers & Behavioral Profiles...")

SEGMENTS = {'retail': 0.80, 'sme': 0.15, 'corporate': 0.05}
customers = []

for i in tqdm(range(NUM_CUSTOMERS), desc="Customers"):
    seg = random_state.choice(list(SEGMENTS.keys()), p=list(SEGMENTS.values()))
    cust_id = f"CUST_{i+100000}"
    
    # Base Demo
    is_retail = seg == 'retail'
    income = int(random_state.lognormal(10 if is_retail else 14, 1.2))
    income = max(3000 if is_retail else 50000, min(200000 if is_retail else 10000000, income))
    
    # Behavioral Profile
    budget_pct = random_state.uniform(0.4, 0.95)
    monthly_outflow = int(income * budget_pct)
    salary_day = random_state.randint(25, 29) if is_retail else None
    
    # Preferred Entities
    fav_merchants = df_merchants.sample(n=random_state.randint(3, 8), random_state=random_state)['merchant_id'].tolist()
    fav_cps = df_counterparties[df_counterparties['cp_type'] == ('Individual' if is_retail else 'Corporate')].sample(n=random_state.randint(1, 4), random_state=random_state)['cp_id'].tolist()
    employer_id = df_employers.sample(n=1, random_state=random_state)['employer_id'].values[0] if is_retail else None
    
    # Volume Baselines
    if is_retail:
        base_txns = random_state.randint(8, 25) if income < 10000 else random_state.randint(15, 60)
    else:
        base_txns = random_state.randint(30, 150)
        
    cust = {
        'customer_id': cust_id,
        'segment': seg,
        'name': fake.name() if is_retail else fake.company(),
        'income_or_turnover': income,
        'budget_pct': budget_pct,
        'expected_monthly_outflow': monthly_outflow,
        'expected_monthly_txns': base_txns,
        'salary_day': salary_day,
        'employer_id': employer_id,
        'fav_merchants': "|".join(fav_merchants),
        'fav_counterparties': "|".join(fav_cps),
        'kyc_risk': random_state.choice(['Low', 'Medium', 'High'], p=[0.7, 0.25, 0.05]),
        'join_date': START_DATE - pd.DateOffset(months=random_state.randint(1, 60))
    }
    
    # Data Quality Imperfections (~2% missing data)
    if random_state.random() < 0.02:
        cust['kyc_risk'] = np.nan
    customers.append(cust)

df_customers = pd.DataFrame(customers)
df_customers.to_csv(f"{OUTPUT_DIR}/customers.csv", index=False)

# =============================================================================
# 3. ACCOUNT GENERATION
# =============================================================================
print("Generating Accounts...")
accounts = []
# Using to_dict('records') is much faster than iterrows()
for cust in df_customers.to_dict('records'):
    n_accs = random_state.choice([1, 2], p=[0.8, 0.2])
    for _ in range(n_accs):
        accounts.append({
            'account_id': f"ACC_{len(accounts)+1000000}",
            'customer_id': cust['customer_id'],
            'currency': random_state.choice(['AED', 'USD', 'EUR'], p=[0.85, 0.10, 0.05]),
            'status': 'Active'
        })
df_accounts = pd.DataFrame(accounts)
df_accounts.to_csv(f"{OUTPUT_DIR}/accounts.csv", index=False)

# Lookups
acc_to_cust = df_accounts.set_index('account_id')['customer_id'].to_dict()
cust_profiles = df_customers.set_index('customer_id').to_dict('index')
merchant_lookup = df_merchants.set_index('merchant_id').to_dict('index')
cp_lookup = df_counterparties.set_index('cp_id').to_dict('index')
employer_lookup = df_employers.set_index('employer_id').to_dict('index')

# =============================================================================
# 4. SCENARIO ENGINE (PRE-GENERATING LINKED TYPOLOGIES)
# =============================================================================
print("Pre-generating linked AML scenarios and False Positives...")

planned_scenarios = {acc: [] for acc in df_accounts['account_id']}
ground_truth_records = []

def add_scenario_txn(acc_id, ts, t_type, amount, d_c, cp_id, cp_name, cp_country, s_id, s_type, role, is_suspicious):
    txn_id = f"TXN_{uuid.uuid4().hex[:10].upper()}"
    txn = {
        'transaction_id': txn_id,
        'account_id': acc_id,
        'timestamp': ts,
        'transaction_type': t_type,
        'amount': amount,
        'direction': d_c,
        'counterparty_id': cp_id,
        'counterparty_name': cp_name,
        'counterparty_country': cp_country,
        'is_scenario': True
    }
    planned_scenarios[acc_id].append(txn)
    
    ground_truth_records.append({
        'transaction_id': txn_id,
        'account_id': acc_id,
        'scenario_id': s_id,
        'scenario_type': s_type,
        'scenario_role': role,
        'is_suspicious': is_suspicious
    })

all_accs = df_accounts['account_id'].tolist()
offshore_cps = df_counterparties[df_counterparties['cp_type'] == 'Offshore']

# 4A. STRUCTURING (Multiple cash deposits summing to > 50k within a week)
for _ in range(int(NUM_CUSTOMERS * 0.015)):
    acc = random_state.choice(all_accs)
    s_id = f"SCEN_STR_{uuid.uuid4().hex[:6]}"
    start_ts = START_DATE + pd.DateOffset(days=random_state.randint(10, 300))
    target_sum = random_state.randint(52000, 75000)
    splits = random_state.randint(4, 7)
    
    for i in range(splits):
        ts = start_ts + pd.DateOffset(days=i, hours=random_state.randint(9, 16))
        amt = target_sum / splits * random_state.uniform(0.8, 1.2)
        add_scenario_txn(acc, ts, 'Cash Deposit', round(amt, 2), 'Credit', None, 'Cash Desk', 'United Arab Emirates', s_id, 'Structuring', 'structuring_deposit', True)

# 4B. RAPID MOVEMENT (Credit in -> Wait hours -> 95% out to Offshore)
for _ in range(int(NUM_CUSTOMERS * 0.01)):
    acc = random_state.choice(all_accs)
    s_id = f"SCEN_RAP_{uuid.uuid4().hex[:6]}"
    ts_in = START_DATE + pd.DateOffset(days=random_state.randint(10, 300), hours=random_state.randint(9, 12))
    amt = random_state.randint(80000, 250000)
    
    add_scenario_txn(acc, ts_in, 'Transfer', amt, 'Credit', None, 'Domestic Partner', 'United Arab Emirates', s_id, 'Rapid Movement', 'source_funding', True)
    
    ts_out = ts_in + pd.DateOffset(hours=random_state.randint(2, 12))
    cp = offshore_cps.sample(1, random_state=random_state).iloc[0]
    add_scenario_txn(acc, ts_out, 'Transfer', round(amt * random_state.uniform(0.9, 0.98), 2), 'Debit', cp['cp_id'], cp['cp_name'], cp['country'], s_id, 'Rapid Movement', 'rapid_outflow', True)

# 4C. LAYERING (Acc A -> Acc B -> Acc C -> Offshore)
for _ in range(int(NUM_CUSTOMERS * 0.005)):
    chain_accs = random_state.choice(all_accs, 3, replace=False)
    s_id = f"SCEN_LAY_{uuid.uuid4().hex[:6]}"
    ts_step1 = START_DATE + pd.DateOffset(days=random_state.randint(10, 300))
    base_amt = random_state.randint(100000, 500000)
    
    add_scenario_txn(chain_accs[0], ts_step1, 'Transfer', base_amt, 'Debit', chain_accs[1], 'Linked Acc B', 'United Arab Emirates', s_id, 'Layering', 'layer_source', True)
    add_scenario_txn(chain_accs[1], ts_step1, 'Transfer', base_amt, 'Credit', chain_accs[0], 'Linked Acc A', 'United Arab Emirates', s_id, 'Layering', 'layer_hop_1', True)
    
    ts_step2 = ts_step1 + pd.DateOffset(days=random_state.randint(1, 3))
    amt_2 = base_amt * random_state.uniform(0.9, 0.98)
    add_scenario_txn(chain_accs[1], ts_step2, 'Transfer', amt_2, 'Debit', chain_accs[2], 'Linked Acc C', 'United Arab Emirates', s_id, 'Layering', 'layer_hop_1', True)
    add_scenario_txn(chain_accs[2], ts_step2, 'Transfer', amt_2, 'Credit', chain_accs[1], 'Linked Acc B', 'United Arab Emirates', s_id, 'Layering', 'layer_hop_2', True)
    
    ts_step3 = ts_step2 + pd.DateOffset(days=random_state.randint(1, 3))
    amt_3 = amt_2 * random_state.uniform(0.9, 0.98)
    cp = offshore_cps.sample(1, random_state=random_state).iloc[0]
    add_scenario_txn(chain_accs[2], ts_step3, 'Transfer', amt_3, 'Debit', cp['cp_id'], cp['cp_name'], cp['country'], s_id, 'Layering', 'layer_destination', True)

# 4D. FALSE POSITIVES (Legitimate large anomalous behavior - e.g. Property Purchase)
for _ in range(int(NUM_CUSTOMERS * 0.02)):
    acc = random_state.choice(all_accs)
    s_id = f"SCEN_FP_{uuid.uuid4().hex[:6]}"
    ts = START_DATE + pd.DateOffset(days=random_state.randint(10, 300))
    amt = random_state.randint(200000, 1500000)
    
    add_scenario_txn(acc, ts - pd.DateOffset(days=2), 'Loan Disbursement', amt, 'Credit', None, 'Internal Bank Loan', 'United Arab Emirates', s_id, 'False Positive', 'legit_funding', False)
    add_scenario_txn(acc, ts, 'Transfer', amt, 'Debit', None, 'Emaar Properties PJSC', 'United Arab Emirates', s_id, 'False Positive', 'property_purchase', False)

# =============================================================================
# 5. CHRONOLOGICAL TRANSACTION GENERATION & BALANCE TRACKING
# =============================================================================
print("Generating sequential transactions and tracking balances...")

out_cols = ['transaction_id', 'account_id', 'customer_id', 'timestamp', 'transaction_type', 
            'channel', 'amount', 'currency', 'amount_aed_equivalent', 'direction', 
            'counterparty_id', 'counterparty_name', 'counterparty_country', 'balance_after']

gt_cols = ['transaction_id', 'account_id', 'scenario_id', 'scenario_type', 'scenario_role', 'is_suspicious']

# Convert accounts dataframe to list of dicts for rapid iteration
acc_records = df_accounts.to_dict('records')

with open(f"{OUTPUT_DIR}/transactions.csv", 'w', newline='', encoding='utf-8') as f_txn, \
     open(f"{OUTPUT_DIR}/ground_truth.csv", 'w', newline='', encoding='utf-8') as f_gt:
         
    writer_txn = csv.DictWriter(f_txn, fieldnames=out_cols)
    writer_gt = csv.DictWriter(f_gt, fieldnames=gt_cols)
    writer_txn.writeheader()
    writer_gt.writeheader()
    
    # Write pre-generated scenario ground truths
    for gt in ground_truth_records:
        writer_gt.writerow(gt)
        
    for acc in tqdm(acc_records, desc="Accounts Processed"):
        acc_id = acc['account_id']
        cust = cust_profiles[acc['customer_id']]
        currency = acc['currency']
        fx = FX_RATES[currency]
        
        is_retail = cust['segment'] == 'retail'
        salary_day_val = cust['salary_day']
        
        # Safely split strings that might be NaN
        fav_merchants = str(cust['fav_merchants']).split('|') if pd.notna(cust.get('fav_merchants')) and cust['fav_merchants'] else []
        fav_cps = str(cust['fav_counterparties']).split('|') if pd.notna(cust.get('fav_counterparties')) and cust['fav_counterparties'] else []
        
        acc_txns = []
        
        current_month = START_DATE
        while current_month < END_DATE:
            # Pandas days_in_month makes this easy and clean
            month_days = current_month.days_in_month
            
            # SALARY (Retail only)
            if is_retail and pd.notna(salary_day_val) and pd.notna(cust.get('employer_id')):
                # SAFELY convert the day to an integer
                clamped_day = int(min(salary_day_val, month_days))
                sal_date = current_month.replace(day=clamped_day)
                sal_ts = sal_date + pd.DateOffset(hours=random_state.randint(8, 12), minutes=random_state.randint(0, 59))
                
                emp = employer_lookup[cust['employer_id']]
                acc_txns.append({
                    'transaction_id': f"TXN_{uuid.uuid4().hex[:10].upper()}",
                    'timestamp': sal_ts,
                    'transaction_type': 'Salary',
                    'channel': 'Automated Clearing',
                    'amount': float(cust['income_or_turnover']),
                    'direction': 'Credit',
                    'counterparty_id': cust['employer_id'],
                    'counterparty_name': emp['employer_name'],
                    'counterparty_country': 'United Arab Emirates',
                    'is_scenario': False
                })
                
                # RENT
                rent_ts = sal_ts + pd.DateOffset(days=random_state.randint(1, 3))
                acc_txns.append({
                    'transaction_id': f"TXN_{uuid.uuid4().hex[:10].upper()}",
                    'timestamp': rent_ts,
                    'transaction_type': 'Rent Payment',
                    'channel': 'Cheque/Transfer',
                    'amount': round(float(cust['income_or_turnover']) * 0.3, 2),
                    'direction': 'Debit',
                    'counterparty_id': None,
                    'counterparty_name': 'Real Estate Agency',
                    'counterparty_country': 'United Arab Emirates',
                    'is_scenario': False
                })

            # SMEs / Corps Inflows
            if not is_retail:
                for _ in range(random_state.randint(2, 6)):
                    rev_ts = current_month + pd.DateOffset(days=random_state.randint(1, month_days), hours=random_state.randint(9, 17))
                    acc_txns.append({
                        'transaction_id': f"TXN_{uuid.uuid4().hex[:10].upper()}",
                        'timestamp': rev_ts,
                        'transaction_type': 'Customer Receipt',
                        'channel': 'Transfer',
                        'amount': round((float(cust['income_or_turnover']) / 12) * random_state.uniform(0.2, 0.6), 2),
                        'direction': 'Credit',
                        'counterparty_id': None,
                        'counterparty_name': 'Corporate Client',
                        'counterparty_country': 'United Arab Emirates',
                        'is_scenario': False
                    })
            
            # DAILY OUTFLOWS
            num_spends = int(cust['expected_monthly_txns'])
            budget_remaining = float(cust['expected_monthly_outflow'])
            
            for _ in range(num_spends):
                if budget_remaining <= 0: break
                
                spend_ts = current_month + pd.DateOffset(days=random_state.randint(1, month_days), hours=random_state.randint(8, 22))
                amt = min(budget_remaining, round(random_state.lognormal(5, 1.5), 2))
                budget_remaining -= amt
                
                is_transfer = random_state.random() < 0.2
                if is_transfer and fav_cps:
                    cp_id = random_state.choice(fav_cps)
                    cp_info = cp_lookup[cp_id]
                    c_name, c_ctry = cp_info['cp_name'], cp_info['country']
                    t_type, chan = 'Transfer', 'Online Banking'
                elif fav_merchants:
                    m_id = random_state.choice(fav_merchants)
                    m_info = merchant_lookup[m_id]
                    cp_id, c_name, c_ctry = m_id, m_info['merchant_name'], m_info['country']
                    t_type, chan = 'POS' if c_ctry == 'United Arab Emirates' else 'E-commerce', 'Card'
                else:
                    cp_id, c_name, c_ctry = None, 'General Merchant', 'United Arab Emirates'
                    t_type, chan = 'POS', 'Card'
                    
                acc_txns.append({
                    'transaction_id': f"TXN_{uuid.uuid4().hex[:10].upper()}",
                    'timestamp': spend_ts,
                    'transaction_type': t_type,
                    'channel': chan,
                    'amount': amt,
                    'direction': 'Debit',
                    'counterparty_id': cp_id,
                    'counterparty_name': c_name,
                    'counterparty_country': c_ctry,
                    'is_scenario': False
                })
                
            current_month += pd.DateOffset(months=1)
            
        # 5B. Inject Pre-planned Scenario Transactions
        acc_txns.extend(planned_scenarios[acc_id])
        
        # 5C. Sort Chronologically & Apply Running Balance
        acc_txns.sort(key=lambda x: x['timestamp'])
        
        running_balance = round(float(cust['income_or_turnover']) * random_state.uniform(1.0, 3.0), 2)
        valid_txns_for_output = []
        
        for t in acc_txns:
            amt = round(t['amount'], 2)
            
            if t['direction'] == 'Credit':
                running_balance += amt
            else:
                if amt > running_balance:
                    if t['is_scenario']:
                        # Auto-fund scenario to prevent it breaking
                        running_balance += (amt * 1.5)
                    else:
                        # Cap normal spending to avoid overdrafts
                        amt = running_balance * 0.95
                        if amt < 5: continue 
                        t['amount'] = amt
                running_balance -= amt
                
            running_balance = round(running_balance, 2)
            
            out_dict = {
                'transaction_id': t['transaction_id'],
                'account_id': acc_id,
                # FIXED: Use acc['customer_id'] instead of cust['customer_id']
                'customer_id': acc['customer_id'],
                'timestamp': t['timestamp'].strftime('%Y-%m-%d %H:%M:%S'),
                'transaction_type': t['transaction_type'],
                'channel': t.get('channel', 'Branch/Wire'),
                'amount': amt,
                'currency': currency,
                # FIXED FX: Multiplying instead of dividing
                'amount_aed_equivalent': round(amt * fx, 2), 
                'direction': t['direction'],
                'counterparty_id': t.get('counterparty_id', ''),
                'counterparty_name': t.get('counterparty_name', ''),
                'counterparty_country': t.get('counterparty_country', ''),
                'balance_after': running_balance
            }
            valid_txns_for_output.append(out_dict)
            
        # Write to disk
        writer_txn.writerows(valid_txns_for_output)

# Save Master tables
df_merchants.to_csv(f"{OUTPUT_DIR}/merchants.csv", index=False)
df_employers.to_csv(f"{OUTPUT_DIR}/employers.csv", index=False)
df_counterparties.to_csv(f"{OUTPUT_DIR}/counterparties.csv", index=False)

print("\n" + "=" * 60)
print("ADVANCED AML DATA GENERATION COMPLETE")
print("=" * 60)
print(f"Output Directory: {OUTPUT_DIR}/")
print("Generated Files:")
print(" - customers.csv (Includes persistent behavioral profiles)")
print(" - accounts.csv")
print(" - merchants.csv, employers.csv, counterparties.csv (Entity resolution tables)")
print(" - transactions.csv (Clean chronological ledger with running balances)")
print(" - ground_truth.csv (Hidden labels mapping TXNs to specific AML Typology Scenarios)")
```
